1. Linear SearchExplanation:


Linear search is the most straightforward searching algorithm. It works by iterating through a sequence (like a list or an array) element by element from the beginning to the end. It compares each element with the target value. If a match is found, it returns the index; if it reaches the end of the list without finding the target, it returns an indicator that the item is not present (usually -1).

It does not require the data to be sorted.Time Complexity: $O(n)$Space Complexity: $O(1)$Core Implementation:

```python
def linear_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1

print(linear_search([10, 2, 3, 4], 6))
```

1. The Standard Search (First Occurrence)
Given an unsorted array of integers, write a function to return the index of the first occurrence of a target integer. If not found, return -1.

```python
def find_first(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1

# Test
print("Q1:", find_first([4, 2, 7, 2, 9], 2)) # Expected output: 1
```

2. The Accumulator (Counting Occurrences)
Given a list of student grades, count how many students scored exactly an 85.

```python
def count_grades(grades, target=85):
    count = 0
    for grade in grades:
        if grade == target:
            count += 1
    return count

# Test
print("Q2:", count_grades([85, 90, 85, 70, 85])) # Expected output: 3
```

3. The Extremes (Finding Minimum)
Given an array of stock prices for the week, find and return the minimum price.

```python
def find_minimum(prices):
    if not prices:
        return None
    min_price = prices[0]
    for price in prices:
        if price < min_price:
            min_price = price
    return min_price

# Test
print("Q3:", find_minimum([120, 115, 130, 110, 125])) # Expected output: 110
```

4. Condition-Based Search (Filtering)
Given a list of mixed integers, return a new list containing only the even numbers that are strictly greater than 10.

```python
def filter_evens(arr):
    result = []
    for num in arr:
        if num > 10 and num % 2 == 0:
            result.append(num)
    return result

# Test
print("Q4:", filter_evens([5, 12, 18, 9, 10, 22])) # Expected output: [12, 18, 22]
```

5. The Last Occurrence (Reverse Search)
Given an array with duplicate elements, find the index of the last occurrence of the target value.

```python
def find_last(arr, target):
    # Iterate backwards using range(start, stop, step)
    for i in range(len(arr) - 1, -1, -1):
        if arr[i] == target:
            return i
    return -1

# Test
print("Q6:", find_last([4, 2, 7, 2, 9], 2)) # Expected output: 3
```

2. Binary SearchExplanation:

Binary search is a highly efficient algorithm that works strictly on sorted arrays. It follows a divide-and-conquer approach. It compares the target value to the middle element of the array. If they are unequal, the half in which the target cannot lie is eliminated, and the search continues on the remaining half until it finds the target or the remaining half is empty.Time Complexity: $O(\log n)$Space Complexity: $O(1)$Core Implementation:

```python
def binary_search(arr, target):
    left, right = 0, len(arr) - 1

    while left <= right:
        mid = (left + right) // 2

        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return -1

binary_search([2, 3, 4, 5,6 , 10] , 2)
```

1. The Standard Search (Exact Match)
Given a sorted array of integers, find the index of a target value.

```python
def standard_binary_search(arr, target):
    left, right = 0, len(arr) - 1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        if arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1

# Test
print("Q1:", standard_binary_search([1, 3, 5, 7, 9], 7)) # Expected output: 3
```

2. Search Insert Position
Given a sorted array and a target value, return the index if the target is found. If not, return the index where it would be if it were inserted in order.

```python
def search_insert(arr, target):
    left, right = 0, len(arr) - 1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        if arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return left # Left pointer rests at the correct insertion point

# Test
print("Q2:", search_insert([1, 3, 5, 6], 2)) # Expected output: 1
```

3. First Position of a Target (Lower Bound)
Given a sorted array with duplicates, find the first occurrence index of a given target value.

```python
def first_occurrence(arr, target):
    left, right = 0, len(arr) - 1
    result = -1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            result = mid
            right = mid - 1 # Keep searching to the left
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return result

# Test
print("Q3:", first_occurrence([1, 2, 2, 2, 5], 2)) # Expected output: 1
```

4. Search in a Rotated Sorted Array
A sorted array is rotated at an unknown pivot (e.g., [4,5,6,7,0,1,2]). Search for a target value and return its index.

```python
def search_rotated(arr, target):
    left, right = 0, len(arr) - 1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid

        # Check if left half is sorted
        if arr[left] <= arr[mid]:
            if arr[left] <= target < arr[mid]:
                right = mid - 1
            else:
                left = mid + 1
        # Otherwise, right half must be sorted
        else:
            if arr[mid] < target <= arr[right]:
                left = mid + 1
            else:
                right = mid - 1
    return -1

# Test
print("Q4:", search_rotated([4, 5, 6, 7, 0, 1, 2], 0)) # Expected output: 4
```

5. Find Peak Element
A peak element is strictly greater than its neighbors. Given an integer array nums, find a peak element and return its index.

```python
def find_peak(nums):
    left, right = 0, len(nums) - 1
    while left < right:
        mid = (left + right) // 2
        # If mid is less than next, peak must be to the right
        if nums[mid] < nums[mid + 1]:
            left = mid + 1
        else:
            # Peak is at mid or to the left
            right = mid
    return left

# Test
print("Q5:", find_peak([1, 2, 3, 1])) # Expected output: 2
```

6. Valid Perfect Square
Given a positive integer num, write a function which returns True if num is a perfect square, else False. Do not use built-in math libraries.

```python
def is_perfect_square(num):
    if num < 1:
        return False
    left, right = 1, num
    while left <= right:
        mid = (left + right) // 2
        squared = mid * mid
        if squared == num:
            return True
        if squared < num:
            left = mid + 1
        else:
            right = mid - 1
    return False

# Test
print("Q6:", is_perfect_square(16)) # Expected output: True
```

3. Jump Search

Explanation:

Jump search operates on sorted arrays. Instead of checking every element, it skips ahead by a fixed number of steps (the optimal step size is $\sqrt{n}$). Once it jumps past an element that is greater than the target, it knows the target must be in the previous block. It then performs a standard linear search backwards (or forwards from the previous step) within that specific block.Time Complexity: $O(\sqrt{n})$Space Complexity: $O(1)$Core Implementation:

```python
import math

def jump_search(arr, target):
    n = len(arr)
    step = int(math.sqrt(n))
    prev = 0

    # Jump blocks
    while prev < n and arr[min(step, n) - 1] < target:
        prev = step
        step += int(math.sqrt(n))
        if prev >= n:
            return -1

    # Linear search in the block
    while prev < n and arr[prev] <= target:
        if arr[prev] == target:
            return prev
        prev += 1

    return -1

jump_search([ 2, 3,4 ,5, 10], 5)
```

1. Custom Jump Size
Modify the algorithm to take a jump_size as an argument instead of calculating the square root of $n$.

```python
def custom_jump(arr, target, jump_size):
    n = len(arr)
    prev = 0
    step = jump_size

    while prev < n and arr[min(step, n) - 1] < target:
        prev = step
        step += jump_size
        if prev >= n: # Added check to prevent out of bounds and infinite loop if target is larger than all elements
            return -1

    for i in range(prev, min(step, n)):
        if arr[i] == target:
            return i
    return -1

custom_jump([1, 2, 3, 1], 1, 2)
```

2. Find First Element Greater Than Target
Use jump search logic to find the index of the first element in a sorted array that is strictly greater than the target.

```python
import math
def first_greater(arr, target):
    n = len(arr)
    step = int(math.sqrt(n))
    prev = 0

    while prev < n and arr[min(step, n) - 1] <= target:
        prev = step
        step += int(math.sqrt(n))
        if prev >= n:
            return -1

    for i in range(prev, min(step, n)):
        if arr[i] > target:
            return i
    return -1

# Test
print("Q4:", first_greater([2, 4, 6, 8, 10], 5)) # Expected output: 2
```

3. Count Target Occurrences in a Block
Given a sorted array with duplicates, use jump search to find the block containing the target, and then count how many times the target appears in the entire array.

```python
import math
def jump_count(arr, target):
    n = len(arr)
    step = int(math.sqrt(n))
    prev = 0

    while prev < n and arr[min(step, n) - 1] < target:
        prev = step
        step += int(math.sqrt(n))
        if prev >= n:
            return 0

    count = 0
    # Search linearly from prev to the end of the array
    for i in range(prev, n):
        if arr[i] == target:
            count += 1
        elif arr[i] > target:
            break # Stop once we exceed target
    return count

# Test
print("Q5:", jump_count([1, 2, 2, 2, 3, 4], 2)) # Expected output: 3
```
