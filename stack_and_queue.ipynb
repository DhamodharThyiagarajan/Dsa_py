**Stack & Queues**

I completely understand why you would want to teach how to add or delete elements in the middle to give your students a comprehensive view of data manipulation. However, I must gently but firmly correct a fundamental misconception here: Stacks and Queues explicitly and mathematically prohibit adding or deleting elements in the middle. If you introduce "add in middle" or "delete in middle" to these structures, they are natively no longer Stacks or Queues—they simply revert to being standard Arrays or Linked Lists.

The entire architectural power of a Stack (LIFO) and a Queue (FIFO) comes from their strict boundary constraints. By aggressively locking down access to only the absolute ends (the top of the stack, or the front/rear of the queue), we guarantee blazing-fast O(1) constant time operations. If you modify the middle, you force the memory to shift, destroying that efficiency and degrading the structure to O(N) linear time.

Here is how you should teach the complete, mathematically strict set of operations (Create, Add, Delete, Peek, and Size) for both structures, emphasizing to your students why the middle is securely off-limits.

Stack: Use the "spring-loaded plate dispenser" analogy. Emphasize that it is physically dangerous to pull from the bottom.

Queue: Use the "grocery store checkout line." Emphasize that cutting the line breaks the fundamental rule of fairness.

**The Complete Stack Operations (LIFO)**


When teaching this, emphasize that all additions (push) and deletions (pop) happen aggressively and exclusively at one boundary: the Top.

````python
class CompleteStack:
    # 1. CREATE: Initialize the pristine container securely
    def __init__(self):
        self.stack = []

    # 2. ADD (Push): Insert strictly at the absolute top boundary
    def push(self, item):
        self.stack.append(item)
        print(f"Pushed: {item}")

    # 3. DELETE (Pop): Violently extract and permanently remove from the top boundary
    def pop(self):
        if self.is_empty():
            return "Stack Underflow: Cannot pop from an empty stack!"
        return self.stack.pop()

    # 4. PEEK (Top): Safely observe the top element without destroying it
    def peek(self):
        if self.is_empty():
            return "Stack is completely empty."
        return self.stack[-1]

    # 5. UTILITY: Check emptiness and size natively
    def is_empty(self):
        return len(self.stack) == 0

    def size(self):
        return len(self.stack)

# ==========================================
# Input / Output Demonstration
# ==========================================
print("--- STACK EXECUTION ---")
# Create
my_stack = CompleteStack()

# Add
my_stack.push("Plate 1 (Bottom)")
my_stack.push("Plate 2 (Middle)")
my_stack.push("Plate 3 (Top)")

# Peek
print(f"Peek at Top: {my_stack.peek()}")
# Expected: Plate 3 (Top)

# Check Size
print(f"Current Size: {my_stack.size()}")
# Expected: 3

# Delete (Pop)
print(f"Popped Element: {my_stack.pop()}")
# Expected: Plate 3 (Top) -> Notice the last one in is the strictly first one out!

print(f"New Top Element: {my_stack.peek()}")
# Expected: Plate 2 (Middle)
````

Evaluate Reverse Polish Notation
**bold text**
Why teach this? This is the definitive "Execution Stack" problem. It shows students exactly how computers evaluate mathematical formulas under the hood. It forces them to extract two items from the top boundary, evaluate them, and push the result back onto the boundary.

Problem Statement: Evaluate an arithmetic expression securely formatted in Reverse Polish Notation. Valid operators are +, -, *, and /.

Input: ["2", "1", "+", "3", "*"]

Output: 9 (Explanation: ((2 + 1) * 3) = 9)

````python
def eval_rpn(tokens):
    stack = []

    for token in tokens:
        # If the token is a mathematical operator, execute extraction
        if token in "+-*/":
            # Violently extract the top TWO elements.
            # Note: The first one popped is the SECOND operand mathematically.
            num2 = stack.pop()
            num1 = stack.pop()

            # Execute the operation and push the aggregate result securely back
            if token == "+": stack.append(num1 + num2)
            elif token == "-": stack.append(num1 - num2)
            elif token == "*": stack.append(num1 * num2)
            elif token == "/": stack.append(int(num1 / num2)) # Truncate toward zero
        else:
            # If it's a raw number, cast it to an integer and push it to the top
            stack.append(int(token))

    # The absolute final remaining element is the flawless answer
    return stack.pop()

# ==========================================
# Input / Output Demonstration
# ==========================================
print("\n--- EVALUATE RPN ---")
tokens1 = ["2", "1", "+", "3", "*"]
print(f"Input: {tokens1} | Output: {eval_rpn(tokens1)}")
# Expected: 9

tokens2 = ["4", "13", "5", "/", "+"]
print(f"Input: {tokens2} | Output: {eval_rpn(tokens2)}")
# Expected: 6 (4 + (13 / 5))
````

**The Complete Queue Operations (FIFO)**


When teaching this, emphasize the dual-boundary system natively implemented via collections.deque. Additions (enqueue) happen strictly at the Rear, and deletions (dequeue) happen strictly at the Front.

````python
from collections import deque

class CompleteQueue:
    # 1. CREATE: Initialize the double-ended queue architecture natively
    def __init__(self):
        self.queue = deque()

    # 2. ADD (Enqueue): Insert incoming elements strictly at the rear boundary
    def enqueue(self, item):
        self.queue.append(item)
        print(f"Enqueued: {item}")

    # 3. DELETE (Dequeue): Extract and remove strictly from the front boundary
    def dequeue(self):
        if self.is_empty():
            return "Queue Underflow: Cannot dequeue from an empty queue!"
        return self.queue.popleft()

    # 4. PEEK FRONT: Safely observe the oldest chronological element
    def peek_front(self):
        if self.is_empty():
            return "Queue is completely empty."
        return self.queue[0]

    # 5. PEEK REAR: Safely observe the newest arrival
    def peek_rear(self):
        if self.is_empty():
            return "Queue is completely empty."
        return self.queue[-1]

    # 6. UTILITY: Check emptiness and size securely
    def is_empty(self):
        return len(self.queue) == 0

    def size(self):
        return len(self.queue)

# ==========================================
# Input / Output Demonstration
# ==========================================
print("\n--- QUEUE EXECUTION ---")
# Create
my_queue = CompleteQueue()

# Add
my_queue.enqueue("Customer 1 (First in line)")
my_queue.enqueue("Customer 2")
my_queue.enqueue("Customer 3 (Last in line)")

# Peek Front and Rear
print(f"Front of Line: {my_queue.peek_front()}")
# Expected: Customer 1 (First in line)
print(f"Back of Line: {my_queue.peek_rear()}")
# Expected: Customer 3 (Last in line)

# Check Size
print(f"Current Size: {my_queue.size()}")
# Expected: 3

# Delete (Dequeue)
print(f"Dequeued Element: {my_queue.dequeue()}")
# Expected: Customer 1 (First in line) -> Notice the first one in is strictly the first one out!

print(f"New Front of Line: {my_queue.peek_front()}")
# Expected: Customer 2
````

# Time Needed to Buy Tickets

This is the ultimate simulation of a continuous Queue. It perfectly demonstrates the action of dequeueing an item from the front, processing it, and aggressively enqueueing it back to the rear if its lifecycle isn't finished.

Problem Statement: There are N people in a line to buy tickets. tickets[i] represents how many tickets person i wants. It takes exactly 1 second to buy a ticket. A person can only buy 1 ticket at a time and must immediately go to the absolute back of the line if they need more. Calculate the exact time it takes for the person sitting at index K to finish buying all their tickets.

Input: tickets = [2, 3, 2], k = 2

Output: 6

````python
from collections import deque

def time_required_to_buy(tickets, k):
    # Initialize a queue to explicitly store the INDICES of the people in line
    queue = deque([i for i in range(len(tickets))])
    time_elapsed = 0

    # Loop continuously until the target person resolves their ticket demand
    while queue:
        # 1. Dequeue the person currently at the absolute front boundary
        current_person = queue.popleft()

        # 2. Process one ticket (takes 1 second)
        tickets[current_person] -= 1
        time_elapsed += 1

        # 3. Check if this was the target person AND they are completely finished
        if tickets[current_person] == 0 and current_person == k:
            return time_elapsed

        # 4. If they still need more tickets, force them to the absolute rear boundary
        if tickets[current_person] > 0:
            queue.append(current_person)

# ==========================================
# Input / Output Demonstration
# ==========================================
print("\n--- BUY TICKETS ---")
tickets1 = [2, 3, 2]
k1 = 2
print(f"Input: tickets={tickets1}, k={k1} | Output: {time_required_to_buy(tickets1, k1)}")
# Expected: 6 (Person 2 gets their tickets at seconds 3 and 6)
````

**How to address the "Middle" question from your students:**

If a student explicitly asks, "How do I delete the second item in the queue?" you should tell them: "You cannot do that without destroying the queue. If you must delete the second item, you must dequeue the first item, pop the second item, and then enqueue the first item back in. If you need arbitrary access to the middle of your dataset, you have architecturally chosen the wrong data structure and should be using a standard Array instead."
