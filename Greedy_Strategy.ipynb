## Part 1: The Greedy Strategy Mindset
Before writing any code, let's clear up a massive point of confusion. When preparing for technical interviews, you often hear terms like *Greedy Algorithm*, *Activity Selection*, and *Fractional Knapsack* thrown around as if they are three competing algorithms. **They are not.**

* **Greedy Algorithm:** The **parent concept**. It is a problem-solving *mindset* or strategy, not a specific block of code.
* **Activity Selection & Fractional Knapsack:** **Child concepts**. They are specific, famous interview problem patterns that we solve *using* the Greedy mindset.

### The Golden Rule of Greedy
Always make the absolute best, most aggressive immediate choice right now, without ever looking backward or worrying about future consequences.

### The Greedy Strategy Flowchart
1. **Identify Constraints & Goals:** What are we limiting (e.g., time, weight)? What are we maximizing (e.g., profit, events)?
2. **The Sorting Phase (Crucial):** Sort the data strictly by a single greedy metric.
3. **The Accumulator:** Initialize variables to track your progress.
4. **The Forward Sweep:** Iterate through the sorted data exactly once. Never reverse a decision.
5. **The Evaluation Gate:** Check if the current item violates constraints. If no, lock it in!

## Part 2: Activity Selection (The Conference Room Problem)

A classic scheduling problem. You have one resource (like a meeting room or a person) and a bunch of events with start and end times. You want to fit in the maximum number of events without overlapping.

f you want to ride the most rollercoasters at a theme park before it closes, you don't pick the longest ride. You greedily pick the ride that gets over with the fastest, so you have time left to run to the next one.
This is a core scheduling problem frequently asked by Google, Meta, and Amazon.

**The Greedy Rule:** Always pick the event that **finishes the earliest**. Why? Because it leaves our schedule as open as possible for future events.

### The Whiteboard Scenario
You have **one** conference room. You want to schedule the absolute maximum number of non-overlapping meetings possible today.
* Meeting A: 10:00 AM to 11:00 AM
* Meeting B: 8:00 AM to 9:30 AM
* Meeting C: 9:00 AM to 12:00 PM
* Meeting D: 11:30 AM to 1:00 PM

### Step-by-Step Greedy Walkthrough
1. **Sort by End Time:** B (9:30), A (11:00), C (12:00), D (1:00)
2. **Meeting B:** Ends 9:30 AM. Room is empty. **[Lock it in]**
3. **Meeting A:** Starts 10:00 AM. Room is free (last meeting ended at 9:30). **[Lock it in]**
4. **Meeting C:** Starts 9:00 AM. We are busy until 11:00 AM. Schedule crash! **[Discard it]**
5. **Meeting D:** Starts 11:30 AM. Room is free (last meeting ended at 11:00). **[Lock it in]**

**Result:** 3 meetings scheduled (B, A, D).

```python
def maximize_meetings(starts, ends):
    # Step 1: Combine starts and ends, then sort STRICTLY by end times
    meetings = list(zip(starts, ends))
    key=lambda x: x[1]
    meetings.sort(key=lambda x: x[1])
    print(meetings)

    # Step 2: Initialize accumulators
    selected_meetings = []
    current_end_time = 0

    # Step 3: Forward Sweep
    for meeting in meetings:
        start_time, end_time = meeting[0], meeting[1]

        # Step 4: Evaluate constraint (Does it start after or when the last one ended?)
        if start_time >= current_end_time:
            selected_meetings.append(meeting)
            current_end_time = end_time # Update the tracking variable

    return selected_meetings

# Test Case
starts = [10.0, 8.0, 9.0, 11.5]
ends = [11.0, 9.5, 12.0, 13.0]
print(maximize_meetings(starts, ends))
```

## Part 3: Fractional Knapsack (The Spice Thief Problem)
This is a resource allocation problem where you have a strict capacity limit, but you are allowed to take *fractions* of items (like liquids, powders, or gold dust).

A resource allocation problem. You have a bag (knapsack) with a strict weight limit, and items that have different weights and monetary values. Crucially, you are allowed to cut or break the items into fractions (like gold dust or liquid). You want to maximize the total money in the bag.

f you are robbing a spice market with a small bag, you don't just grab the heaviest barrel. You calculate which spice is worth the most per ounce (like saffron), and you greedily fill your bag with that first.

**The Greedy Rule:** Always pick the item with the highest **value-to-weight ratio** (price per kg).

### The Whiteboard Scenario
A thief breaks into a spice market. Their bag can hold exactly **50 kg**. They want to maximize their profit.
* Saffron: 10 kg total, worth \$600
* Vanilla: 20 kg total, worth \$1000
* Cinnamon: 30 kg total, worth \$1200

### Step-by-Step Greedy Walkthrough
1. **Calculate Ratios & Sort:**
   * Saffron: \$600 / 10 kg = **\$60/kg**
   * Vanilla: \$1000 / 20 kg = **\$50/kg**
   * Cinnamon: \$1200 / 30 kg = **\$40/kg**
2. **Pack Saffron:** Takes 10 kg. Fits easily. *(Remaining: 40 kg | Profit: \$600)*
3. **Pack Vanilla:** Takes 20 kg. Fits easily. *(Remaining: 20 kg | Profit: \$1600)*
4. **Pack Cinnamon:** Takes 30 kg. It doesn't fit completely!
   * *The Fractional Move:* Take exactly the 20 kg we have space for. (20 kg * \$40 = \$800).

**Result:** Total maximum profit = \$2400.

```python
def fractional_knapsack(capacity, items):
    # Items are provided as list of tuples: (value, weight)
    # Step 1: Sort items strictly descending by value-to-weight ratio
    items.sort(key=lambda x: x[0] / x[1], reverse=True)

    # Step 2: Initialize accumulator
    total_value = 0.0

    # Step 3: Forward Sweep
    for value, weight in items:
        # Step 4: Evaluate constraint
        if capacity >= weight:
            # Take the whole item
            capacity -= weight
            total_value += value
        else:
            # Take a fraction to fill the exact remaining capacity
            fraction = capacity / weight
            total_value += value * fraction
            break # Bag is completely full

    return total_value

# Test Case
items = [(600, 10), (1000, 20), (1200, 30)]
capacity = 50
print("Maximum Profit:", fractional_knapsack(capacity, items))
```

## **Best Time to Buy and Sell Stock II**.

**The Scenario:** You have an array of daily stock prices. You can buy and sell on the exact same day, and you can make as many transactions as you want (but you can only hold one share at a time). Maximize your profit.

**The Greedy Rule:** If tomorrow's price is strictly higher than today's price, "buy" today and "sell" tomorrow. Greedily capture *every single upward movement* in the graph.

### The Whiteboard Scenario
Prices: `[7, 1, 5, 3, 6, 4]`

### Step-by-Step Greedy Walkthrough
1. **Day 1 to 2 (7 to 1):** Price dropped. Do nothing.
2. **Day 2 to 3 (1 to 5):** Price increased! Greedily capture this. Profit += (5 - 1) = 4.
3. **Day 3 to 4 (5 to 3):** Price dropped. Do nothing.
4. **Day 4 to 5 (3 to 6):** Price increased! Greedily capture this. Profit += (6 - 3) = 3.
5. **Day 5 to 6 (6 to 4):** Price dropped. Do nothing.

**Result:** Total maximum profit = 4 + 3 = 7.

```python
def max_profit_stock(prices):
    # Step 1: Initialize accumulator
    max_profit = 0

    # Step 2: Forward Sweep starting from day 2 (index 1)
    for i in range(1, len(prices)):

        # Step 3: Evaluate constraint (Is today's price higher than yesterday's?)
        if prices[i] > prices[i - 1]:
            # Greedily lock in the profit
            max_profit += (prices[i] - prices[i - 1])

    return max_profit

prices = [7, 1, 5, 3, 6, 4]
print("Maximum Stock Profit:", max_profit_stock(prices))
```

## Practice Set 1: Activity Selection & Scheduling (Intervals)
These problems all revolve around the core concept of sorting intervals by a specific boundary (usually the end time or start time) and sweeping forward to resolve overlaps.

### 1. Merge Intervals
* **Problem:** Given an array of intervals, merge all overlapping intervals into one contiguous interval.
* **Input:** `[[1, 3], [2, 6], [8, 10], [15, 18]]`
* **Output:** `[[1, 6], [8, 10], [15, 18]]`
* **Greedy Logic:** Sort intervals strictly by **start time**. Create an empty result list. Iterate forward. If the current interval's start time is $\le$ the previous interval's end time, they overlap. Greedily extend the previous interval's end time to the maximum of the two.

### 2. Non-overlapping Intervals
* **Problem:** Find the minimum number of intervals you must remove to make the remaining intervals non-overlapping.
* **Input:** `[[1, 2], [2, 3], [3, 4], [1, 3]]`
* **Output:** `1` (remove `[1, 3]`)
* **Greedy Logic:** Sort intervals strictly by **end time**. Keep a tracker for the `current_end`. If the next interval starts *before* `current_end`, it is a violation—greedily count it as a removal. Otherwise, update `current_end` to this new interval's end time.

### 3. Meeting Rooms II
* **Problem:** Given an array of meeting time intervals, find the absolute minimum number of conference rooms required.
* **Input:** `[[0, 30], [5, 10], [15, 20]]`
* **Output:** `2`
* **Greedy Logic:** Separate start times and end times into two distinct arrays. Sort both ascending. Use two pointers. If a meeting starts before the earliest ending meeting finishes, you must allocate a new room. If it starts after, a room has freed up, so advance the end-time pointer.

### 4. Minimum Number of Arrows to Burst Balloons
* **Problem:** Balloons are mapped by horizontal intervals. An arrow shot vertically bursts all balloons it passes through. Find the minimum arrows required.
* **Input:** `[[10, 16], [2, 8], [1, 6], [7, 12]]`
* **Output:** `2`
* **Greedy Logic:** Sort balloons strictly by **end coordinates**. Shoot your first arrow at the exact end coordinate of the first balloon. Ignore any subsequent balloons that start *before or exactly at* this arrow's position. When a balloon finally starts *after* the arrow, you must shoot a new arrow at its end coordinate.

### 5. Insert Interval
* **Problem:** Insert a new interval into a sorted array of non-overlapping intervals and merge if necessary.
* **Input:** `intervals = [[1, 3], [6, 9]]`, `newInterval = [2, 5]`
* **Output:** `[[1, 5], [6, 9]]`
* **Greedy Logic:** Iterate through the sorted intervals. Greedily add all intervals that end *before* the new interval starts. When an overlap occurs, greedily merge them by taking the `min` of the starts and `max` of the ends. Finally, add all remaining intervals.

## Practice Set 2: Fractional Knapsack & Capacity Logic
These problems disguise the "value-to-weight ratio" logic. You must identify a strict capacity and a greedy metric to consume that capacity optimally.

### 6. Maximum Units on a Truck
* **Problem:** You have a strict truck capacity (number of boxes) and a 2D array of `[numberOfBoxes, unitsPerBox]`. Maximize the total units.
* **Input:** `boxTypes = [[1, 3], [2, 2], [3, 1]]`, `truckSize = 4`
* **Output:** `8`
* **Greedy Logic:** Sort the box types descending strictly by **units per box** (this is your ratio). Greedily take as many boxes of the highest unit count as possible. Take 1 box of 3 units, then 2 boxes of 2 units, then 1 box of 1 unit. Total: 3 + 4 + 1 = 8.

### 7. Assign Cookies
* **Problem:** Array `g` is the minimum greed factor of children. Array `s` is the sizes of available cookies. Maximize the number of content children.
* **Input:** `g = [1, 2, 3]`, `s = [1, 1]`
* **Output:** `1`
* **Greedy Logic:** Sort both arrays ascending. This is a two-pointer greedy sweep. Greedily try to satisfy the least greedy child with the absolute smallest cookie that works. If the cookie is too small, skip it. If it fits, increment both the child and cookie pointers.

### 8. Task Scheduler
* **Problem:** Given an array of CPU tasks and a cooldown `n`, find the minimum time to finish all tasks. Identical tasks must be separated by `n` intervals.
* **Input:** `tasks = ["A", "A", "A", "B", "B", "B"]`, `n = 2`
* **Output:** `8` (A -> B -> idle -> A -> B -> idle -> A -> B)
* **Greedy Logic:** Calculate the frequencies of each task. Greedily schedule the most frequent task first, as it dictates the maximum number of mandatory idle slots. Fill the resulting idle slots with the remaining less-frequent tasks.

### 9. IPO (Maximize Capital)
* **Problem:** You have initial capital `k` and arrays of `profits` and `capital` required for various projects. Select up to `n` projects to maximize your final capital.
* **Input:** `k = 2`, `w = 0`, `profits = [1, 2, 3]`, `capital = [0, 1, 1]`
* **Output:** `4`
* **Greedy Logic:** Push all projects you can currently afford into a Max-Heap based on their profit. Greedily pop the most profitable project, add its profit to your capital, and then check if this new capital allows you to afford new projects. Repeat up to `n` times.

### 10. Minimum Number of Refueling Stops
* **Problem:** Find the minimum number of gas stations you must stop at to reach a target destination.
* **Input:** `target = 100`, `startFuel = 10`, `stations = [[10, 60], [20, 30], [30, 30], [60, 40]]`
* **Output:** `2`
* **Greedy Logic:** Drive past stations without stopping, but push their fuel capacities into a Max-Heap. When you run out of fuel, greedily pop the largest fuel capacity from the heap (retroactively "stopping" at the best station you passed) to keep going.

## Practice Set 3: Array Sweeps & State Tracking
These problems require you to sweep forward through an array exactly once while tracking a dynamic "state" (like max reach, profit, or cash register bills).

### 11. Best Time to Buy and Sell Stock II
* **Problem:** You can buy and sell stock on the same day, making unlimited transactions. Maximize total profit.
* **Input:** `[7, 1, 5, 3, 6, 4]`
* **Output:** `7`
* **Greedy Logic:** Sweep forward. If tomorrow's price is higher than today's, "buy" today and "sell" tomorrow. Greedily capture every single upward price movement. (5 - 1) + (6 - 3) = 7.

### 12. Jump Game
* **Problem:** An array of integers represents your max jump length from that index. Can you reach the final index?
* **Input:** `[2, 3, 1, 1, 4]`
* **Output:** `True`
* **Greedy Logic:** Maintain a variable `farthest_reach` initialized to 0. Sweep forward. At each step, update `farthest_reach = max(farthest_reach, i + nums[i])`. If your current index `i` is strictly greater than `farthest_reach`, you are trapped.

### 13. Jump Game II
* **Problem:** Find the absolute minimum number of jumps required to reach the last index.
* **Input:** `[2, 3, 1, 1, 4]`
* **Output:** `2` (Jump 1 step from index 0 to 1, then 3 steps to the end)
* **Greedy Logic:** Track `current_jump_end` and `farthest_reach`. Iterate forward. Continually update `farthest_reach`. When your loop hits the `current_jump_end`, you are mathematically forced to make a jump. Increment your jump counter and update `current_jump_end = farthest_reach`.

### 14. Lemonade Change
* **Problem:** Customers pay for a $5 lemonade with $5, $10, or $20 bills. Can you perfectly provide exact change to everyone in order?
* **Input:** `[5, 5, 5, 10, 20]`
* **Output:** `True`
* **Greedy Logic:** Track your $5 and $10 bills. When a customer gives a $20, you must give $15 in change. The **greedy choice** is to always give one $10 and one $5 if possible. You aggressively hoard your $5 bills because they are mathematically required to make change for $10 bills later.

### 15. Gas Station
* **Problem:** Find the starting gas station index that allows you to complete a full circular route without your tank hitting negative fuel.
* **Input:** `gas = [1, 2, 3, 4, 5]`, `cost = [3, 4, 5, 1, 2]`
* **Output:** `3` (Index 3)
* **Greedy Logic:** Track `total_tank` and `current_tank`. Sweep forward calculating `gas[i] - cost[i]`. If `current_tank` drops below 0, you cannot reach the current station from your starting point. Greedily reset your starting point to `i + 1` and reset `current_tank` to 0.
