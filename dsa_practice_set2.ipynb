## Easy 1 — First Unique Character in a String
Find the index of the first non-repeating character. Return -1 if none.

**Example:**
```
Input:  s = "leetcode"
Output: 0

Input:  s = "aabb"
Output: -1
```

````python
from collections import Counter

def first_unique_char(s: str) -> int:
    freq = Counter(s)
    for i, ch in enumerate(s):
        if freq[ch] == 1:
            return i
    return -1

assert first_unique_char('leetcode') == 0
assert first_unique_char('aabb') == -1
assert first_unique_char('z') == 0
print('All tests passed ✅')
````

## Easy 2 — Find the Majority Element
Given an array of size n, find the element that appears more than n/2 times. Use Boyer-Moore voting (O(n) time, O(1) space).

**Example:**
```
Input:  nums = [3,2,3]
Output: 3

Input:  nums = [2,2,1,1,2,2,2]
Output: 2
```

````python
def majority_element(nums: list) -> int:
    candidate, count = None, 0
    for n in nums:
        if count == 0:
            candidate = n
        count += 1 if n == candidate else -1
    return candidate

assert majority_element([3,2,3]) == 3
assert majority_element([2,2,1,1,2,2,2]) == 2
assert majority_element([1]) == 1
print('All tests passed ✅')
````

## Easy 4 — Check If Two String Halves Have Equal Vowel Count

Given an even-length string, split it into two halves. Return `True` if both halves have the same number of vowels.

**Example:**
```
Input:  s = "book"
Output: True   # "bo" has 1 vowel, "ok" has 1 vowel

Input:  s = "textbook"
Output: False
```

````python
def halves_alike(s: str) -> bool:
    vowels = set('aeiouAEIOU')
    mid = len(s) // 2
    return sum(c in vowels for c in s[:mid]) == sum(c in vowels for c in s[mid:])

assert halves_alike("book") == True
assert halves_alike("textbook") == False
print('All tests passed ✅')
````

## Easy 5 — Count Inversions in Array

Given an array, count the number of pairs `(i, j)` where `i < j` and `nums[i] > nums[j]` (brute-force O(n²) is acceptable for easy constraints).

**Example:**
```
Input:  nums = [3, 1, 2]
Output: 2   # (3,1) and (3,2)

Input:  nums = [1, 2, 3]
Output: 0
```

````python
def count_inversions(nums: list) -> int:
    count = 0
    n = len(nums)
    for i in range(n):
        for j in range(i + 1, n):
            if nums[i] > nums[j]:
                count += 1
    return count

assert count_inversions([3, 1, 2]) == 2
assert count_inversions([1, 2, 3]) == 0
print('All tests passed ✅')
````

## Easy 6 — Minimum Cost to Split Bill Among Friends

Three friends A, B, C each paid some amount. Find who owes whom to settle the bill equally. Return a list of `"X pays Y amount"` strings (only non-zero transactions).

**Example:**
```
Input:  a=30, b=60, c=90
Output: ["A pays B 20", "A pays C 10"]

Input:  a=60, b=60, c=60
Output: []
```

````python
def settle_bill(a: int, b: int, c: int) -> list:
    total = a + b + c
    share = total // 3
    balances = {'A': a - share, 'B': b - share, 'C': c - share}
    result = []
    creditors = sorted([(v, k) for k, v in balances.items() if v > 0], reverse=True)
    debtors  = sorted([(-v, k) for k, v in balances.items() if v < 0], reverse=True)
    i, j = 0, 0
    while i < len(creditors) and j < len(debtors):
        credit, cname = creditors[i]
        debt, dname   = debtors[j]
        amount = min(credit, debt)
        result.append(f"{dname} pays {cname} {amount}")
        creditors[i] = (credit - amount, cname)
        debtors[j]   = (debt - amount, dname)
        if creditors[i][0] == 0: i += 1
        if debtors[j][0] == 0:   j += 1
    return result

assert settle_bill(60, 60, 60) == []
r = settle_bill(30, 60, 90)
assert len(r) > 0
print('All tests passed ✅')
````

## Medium 1 — Partition String Into Minimum Segments

Partition a string so each character appears in at most one segment. Return list of segment lengths (greedy: each segment extends to the last occurrence of every character inside it).

**Example:**
```
Input:  s = "ababcbacadefegdehijhklij"
Output: [9, 7, 8]

Input:  s = "eccbbbbdec"
Output: [10]
```

````python
def partition_labels(s: str) -> list:
    last = {c: i for i, c in enumerate(s)}
    result = []
    start = end = 0
    for i, c in enumerate(s):
        end = max(end, last[c])
        if i == end:
            result.append(end - start + 1)
            start = i + 1
    return result

assert partition_labels("ababcbacadefegdehijhklij") == [9, 7, 8]
assert partition_labels("eccbbbbdec") == [10]
print('All tests passed ✅')
````

## Medium 2 — Count Subarrays with Product Below Threshold

Given a positive integer array and threshold `k`, return count of contiguous subarrays whose product is strictly less than `k`.

**Example:**
```
Input:  nums = [10,5,2,6], k = 100
Output: 8
```

````python
def num_subarray_product_less_than_k(nums: list, k: int) -> int:
    if k <= 1:
        return 0
    product = count = 0
    product = 1
    left = 0
    for right in range(len(nums)):
        product *= nums[right]
        while product >= k:
            product //= nums[left]
            left += 1
        count += right - left + 1
    return count

assert num_subarray_product_less_than_k([10,5,2,6], 100) == 8
assert num_subarray_product_less_than_k([1,2,3], 0) == 0
print('All tests passed ✅')
````

## Medium 4 — Zero Out Rows and Columns

Given an m×n matrix, if any cell is 0, set its entire row and column to 0 in-place.

**Example:**
```
Input:  [[1,1,1],[1,0,1],[1,1,1]]
Output: [[1,0,1],[0,0,0],[1,0,1]]
```

````python
def set_zeroes(matrix: list) -> None:
    rows, cols = set(), set()
    m, n = len(matrix), len(matrix[0])
    for i in range(m):
        for j in range(n):
            if matrix[i][j] == 0:
                rows.add(i)
                cols.add(j)
    for i in range(m):
        for j in range(n):
            if i in rows or j in cols:
                matrix[i][j] = 0

m = [[1,1,1],[1,0,1],[1,1,1]]
set_zeroes(m)
assert m == [[1,0,1],[0,0,0],[1,0,1]]

m2 = [[0,1,2,0],[3,4,5,2],[1,3,1,5]]
set_zeroes(m2)
assert m2 == [[0,0,0,0],[0,4,5,0],[0,3,1,0]]
print('All tests passed ✅')
````

## Medium 6 — Unique Paths in Grid with Obstacles

An m×n grid has obstacles (`1`). Count unique paths from top-left to bottom-right moving only right or down.

**Example:**
```
Input:  obstacleGrid = [[0,0,0],[0,1,0],[0,0,0]]
Output: 2

Input:  obstacleGrid = [[0,1],[0,0]]
Output: 1
```

````python
def unique_paths_with_obstacles(grid: list) -> int:
    m, n = len(grid), len(grid[0])
    if grid[0][0] == 1 or grid[m-1][n-1] == 1:
        return 0
    dp = [[0]*n for _ in range(m)]
    dp[0][0] = 1
    for i in range(m):
        for j in range(n):
            if i == 0 and j == 0:
                continue
            if grid[i][j] == 1:
                dp[i][j] = 0
            else:
                dp[i][j] = (dp[i-1][j] if i > 0 else 0) + (dp[i][j-1] if j > 0 else 0)
    return dp[m-1][n-1]

assert unique_paths_with_obstacles([[0,0,0],[0,1,0],[0,0,0]]) == 2
assert unique_paths_with_obstacles([[0,1],[0,0]]) == 1
print('All tests passed ✅')
````

## Medium 7 — All Root-to-Leaf Paths with Target Sum

Given a binary tree and target sum, return all root-to-leaf paths whose node values sum to the target.

**Example:**
```
Tree:       5
           / \
          4   8
         /   / \
        11  13   4
       /  \    / \
      7    2  5   1

Input:  targetSum = 22
Output: [[5,4,11,2],[5,8,4,5]]
```

````python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def path_sum(root: TreeNode, targetSum: int) -> list:
    result = []

    def dfs(node, remaining, path):
        if not node:
            return
        path.append(node.val)
        if not node.left and not node.right and remaining == node.val:
            result.append(path[:])
        dfs(node.left,  remaining - node.val, path)
        dfs(node.right, remaining - node.val, path)
        path.pop()

    dfs(root, targetSum, [])
    return result

# Build example tree
root = TreeNode(5,
    TreeNode(4, TreeNode(11, TreeNode(7), TreeNode(2))),
    TreeNode(8, TreeNode(13), TreeNode(4, TreeNode(5), TreeNode(1)))
)
assert sorted(path_sum(root, 22)) == sorted([[5,4,11,2],[5,8,4,5]])
assert path_sum(None, 0) == []
assert path_sum(TreeNode(1, TreeNode(2)), 1) == []
print('All tests passed ✅')
````

## Medium 8 — Validate Binary Search Tree

Return `True` if the binary tree is a valid BST (left subtree < node < right subtree, strictly, for all nodes).

**Example:**
```
    2
   / \
  1   3     →  True

    5
   / \
  1   4
     / \
    3   6  →  False
```

````python
def is_valid_bst(root) -> bool:
    def validate(node, lo, hi):
        if not node:
            return True
        if not (lo < node.val < hi):
            return False
        return validate(node.left, lo, node.val) and validate(node.right, node.val, hi)
    return validate(root, float('-inf'), float('inf'))

# Valid BST: 2->1,3
valid = TreeNode(2, TreeNode(1), TreeNode(3))
assert is_valid_bst(valid) == True
# Invalid: 5->(1, 4->(3,6))
invalid = TreeNode(5, TreeNode(1), TreeNode(4, TreeNode(3), TreeNode(6)))
assert is_valid_bst(invalid) == False
print('All tests passed ✅')
````

## Medium 9 — Count Subarrays Equal to Maximum Sum

Find the maximum subarray sum (Kadane's), then count how many subarrays achieve exactly that sum.

**Example:**
```
Input:  nums = [1,-1,1,1,-1,1]
Output: (2, 4)   # max_sum=2, count=4

Input:  nums = [3]
Output: (3, 1)
```

````python
def count_max_subarray(nums: list):
    # Step 1: Find max sum (Kadane)
    max_sum = cur = nums[0]
    for x in nums[1:]:
        cur = max(x, cur + x)
        max_sum = max(max_sum, cur)
    # Step 2: Count subarrays with that sum using prefix sums
    from collections import defaultdict
    prefix_count = defaultdict(int)
    prefix_count[0] = 1
    running = 0
    count = 0
    for x in nums:
        running += x
        count += prefix_count[running - max_sum]
        prefix_count[running] += 1
    return max_sum, count

assert count_max_subarray([1,-1,1,1,-1,1]) == (2, 4)
assert count_max_subarray([3]) == (3, 1)
print('All tests passed ✅')
````

## Medium 10 — Remove Duplicates from Sorted Linked List

Given a sorted linked list, delete all nodes with duplicate values, leaving only distinct values.

**Example:**
```
Input:  1 -> 1 -> 2 -> 3 -> 3
Output: 1 -> 2 -> 3

Input:  1 -> 1 -> 1
Output: 1
```

````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def to_list(head):
    out = []
    while head:
        out.append(head.val)
        head = head.next
    return out

def make_list(vals):
    dummy = ListNode()
    cur = dummy
    for v in vals:
        cur.next = ListNode(v)
        cur = cur.next
    return dummy.next

def delete_duplicates(head: ListNode) -> ListNode:
    cur = head
    while cur and cur.next:
        if cur.val == cur.next.val:
            cur.next = cur.next.next
        else:
            cur = cur.next
    return head

assert to_list(delete_duplicates(make_list([1,1,2,3,3]))) == [1,2,3]
assert to_list(delete_duplicates(make_list([1,1,1]))) == [1]
assert to_list(delete_duplicates(make_list([1,2,3]))) == [1,2,3]
print('All tests passed ✅')
````

## Medium 11 — Check Palindrome Linked List

Return `True` if the linked list is a palindrome. Solve in O(n) time and O(1) space.

**Example:**
```
Input:  1 -> 2 -> 2 -> 1
Output: True

Input:  1 -> 2
Output: False
```

````python
def is_palindrome(head: ListNode) -> bool:
    # Find middle
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
    # Reverse second half
    prev = None
    while slow:
        nxt = slow.next
        slow.next = prev
        prev = slow
        slow = nxt
    # Compare
    left, right = head, prev
    while right:
        if left.val != right.val:
            return False
        left = left.next
        right = right.next
    return True

assert is_palindrome(make_list([1,2,2,1])) == True
assert is_palindrome(make_list([1,2])) == False
assert is_palindrome(make_list([1,2,1])) == True
assert is_palindrome(make_list([1])) == True
print('All tests passed ✅')
````

## Medium 12 — Reverse a Linked List

Reverse a singly-linked list iteratively and return the new head.

**Example:**
```
Input:  1 -> 2 -> 3 -> 4 -> 5
Output: 5 -> 4 -> 3 -> 2 -> 1
```

````python
def reverse_list(head: ListNode) -> ListNode:
    prev = None
    cur = head
    while cur:
        nxt = cur.next
        cur.next = prev
        prev = cur
        cur = nxt
    return prev

assert to_list(reverse_list(make_list([1,2,3,4,5]))) == [5,4,3,2,1]
assert to_list(reverse_list(make_list([1,2]))) == [2,1]
assert to_list(reverse_list(make_list([1]))) == [1]
print('All tests passed ✅')
````

## Medium 13 — Evaluate Postfix (RPN) Expression

Evaluate a Reverse Polish Notation expression. Operators: `+`, `-`, `*`, `/` (integer division truncated toward zero).

**Example:**
```
Input:  tokens = ["2","1","+","3","*"]
Output: 9

Input:  tokens = ["4","13","5","/","+"]
Output: 6
```

````python
import operator

def eval_rpn(tokens: list) -> int:
    ops = {'+': operator.add, '-': operator.sub, '*': operator.mul,
           '/': lambda a, b: int(a / b)}
    stack = []
    for t in tokens:
        if t in ops:
            b, a = stack.pop(), stack.pop()
            stack.append(ops[t](a, b))
        else:
            stack.append(int(t))
    return stack[0]

assert eval_rpn(["2","1","+","3","*"]) == 9
assert eval_rpn(["4","13","5","/","+"]) == 6
assert eval_rpn(["10","6","9","3","+","-11","*","/","*","17","+","5","+"]) == 22
assert eval_rpn(["18"]) == 18
print('All tests passed ✅')
````

## Medium 14 — Validate Bracket String

Given a string of `(`, `)`, `{`, `}`, `[`, `]`, return `True` if it is correctly matched and nested.

**Example:**
```
Input:  s = "()[]{}"
Output: True

Input:  s = "(]"
Output: False

Input:  s = "{[]}"
Output: True
```

````python
def is_valid_parens(s: str) -> bool:
    match = {')':'(', ']':'[', '}':'{'}
    stack = []
    for ch in s:
        if ch in match:
            if not stack or stack[-1] != match[ch]:
                return False
            stack.pop()
        else:
            stack.append(ch)
    return not stack

assert is_valid_parens("()[]{}") == True
assert is_valid_parens("(]") == False
assert is_valid_parens("{[]}") == True
print('All tests passed ✅')
````

## Medium 16 — Kth Largest Element in an Array
Find the k-th largest element using a min-heap of size k.

**Example:**
```
Input:  nums = [3,2,1,5,6,4], k = 2
Output: 5

Input:  nums = [3,2,3,1,2,4,5,5,6], k = 4
Output: 4
```

````python
import heapq

def find_kth_largest(nums: list, k: int) -> int:
    heap = []
    for n in nums:
        heapq.heappush(heap, n)
        if len(heap) > k:
            heapq.heappop(heap)
    return heap[0]

assert find_kth_largest([3,2,1,5,6,4], 2) == 5
assert find_kth_largest([3,2,3,1,2,4,5,5,6], 4) == 4
print('All tests passed ✅')
````

## Medium 17 — Longest Consecutive Sequence
Find the length of the longest consecutive elements sequence in O(n).

**Example:**
```
Input:  nums = [100,4,200,1,3,2]
Output: 4

Input:  nums = [0,3,7,2,5,8,4,6,0,1]
Output: 9
```

````python
def longest_consecutive(nums: list) -> int:
    s = set(nums)
    best = 0
    for n in s:
        if n - 1 not in s:
            cur, length = n, 1
            while cur + 1 in s:
                cur += 1; length += 1
            best = max(best, length)
    return best

assert longest_consecutive([100,4,200,1,3,2]) == 4
assert longest_consecutive([0,3,7,2,5,8,4,6,0,1]) == 9
assert longest_consecutive([]) == 0
print('All tests passed ✅')
````

## Medium 18 — Ratio Queries on a Graph (Evaluate Division)

Given equations like `A/B = k`, build a weighted graph and answer queries `X/Y` using BFS/DFS. Return `-1.0` for unknown variables.

**Example:**
```
equations = [["a","b"],["b","c"]], values = [2.0, 3.0]
queries   = [["a","c"],["b","a"],["x","y"]]
Output: [6.0, 0.5, -1.0]
```

````python
from collections import defaultdict, deque

def calc_equation(equations, values, queries):
    graph = defaultdict(dict)
    for (a, b), v in zip(equations, values):
        graph[a][b] = v
        graph[b][a] = 1 / v

    def bfs(src, dst):
        if src not in graph or dst not in graph:
            return -1.0
        if src == dst:
            return 1.0
        visited = set()
        q = deque([(src, 1.0)])
        while q:
            node, product = q.popleft()
            if node == dst:
                return product
            visited.add(node)
            for neighbour, weight in graph[node].items():
                if neighbour not in visited:
                    q.append((neighbour, product * weight))
        return -1.0

    return [bfs(s, d) for s, d in queries]

res = calc_equation([["a","b"],["b","c"]], [2.0,3.0], [["a","c"],["b","a"],["x","y"]])
assert abs(res[0] - 6.0) < 1e-9
assert abs(res[1] - 0.5) < 1e-9
assert res[2] == -1.0
print('All tests passed ✅')
````

## Medium 19 — Rearrange Deck for Ascending Reveal Order

Arrange a deck so that repeatedly revealing the top card and moving the next to the bottom reveals cards in ascending order.

**Example:**
```
Input:  cards = [17,13,11,2,3,5,7]
Output: [2,13,3,11,5,17,7]
```

````python
from collections import deque

def reveal_increasing(deck: list) -> list:
    n = len(deck)
    deck.sort()
    indices = deque(range(n))
    result = [0] * n
    for card in deck:
        result[indices.popleft()] = card
        if indices:
            indices.append(indices.popleft())
    return result

assert reveal_increasing([17,13,11,2,3,5,7]) == [2,13,3,11,5,17,7]
assert reveal_increasing([1,1000]) == [1,1000]
print('All tests passed ✅')
````

## Medium 20 — Car Fleet Count

Cars on a road each have a position and speed. Cars that catch up to a slower car ahead form a fleet. Return the number of fleets reaching the target.

**Example:**
```
Input:  target=12, position=[10,8,0,5,3], speed=[2,4,1,1,3]
Output: 3
```

````python
def car_fleet(target: int, position: list, speed: list) -> int:
    stack = []
    for pos, spd in sorted(zip(position, speed), reverse=True):
        t = (target - pos) / spd
        if not stack or t > stack[-1]:
            stack.append(t)
    return len(stack)

assert car_fleet(12, [10,8,0,5,3], [2,4,1,1,3]) == 3
assert car_fleet(10, [3], [3]) == 1
assert car_fleet(100, [0,2,4], [4,2,1]) == 1
print('All tests passed ✅')
````

## Medium 21 — Minimum Intervals to Complete All Tasks

Given a list of tasks (letters) and cooldown `n`, find the minimum CPU intervals needed. Idle slots are inserted when a task must wait.

**Example:**
```
Input:  tasks = ['A','A','A','B','B','B'], n = 2
Output: 8

Input:  tasks = ['A','A','A','B','B','B'], n = 0
Output: 6
```

````python
from collections import Counter

def least_interval(tasks: list, n: int) -> int:
    freq = Counter(tasks)
    max_freq = max(freq.values())
    max_count = sum(1 for v in freq.values() if v == max_freq)
    return max(len(tasks), (max_freq - 1) * (n + 1) + max_count)

assert least_interval(['A','A','A','B','B','B'], 2) == 8
assert least_interval(['A','A','A','B','B','B'], 0) == 6
print('All tests passed ✅')
````

## Hard 1 — Count All Nodes in Complete Binary Tree

Given a complete binary tree, count its nodes in O(log²n) time by exploiting the complete property (compare left and right heights).

**Example:**
```
Tree (complete, 6 nodes):
        1
       / \
      2   3
     / \ /
    4  5 6
Output: 6
```

````python
def count_nodes(root: TreeNode) -> int:
    if not root:
        return 0

    def left_height(node):
        h = 0
        while node:
            h += 1
            node = node.left
        return h

    def right_height(node):
        h = 0
        while node:
            h += 1
            node = node.right
        return h

    lh = left_height(root)
    rh = right_height(root)
    if lh == rh:
        return (1 << lh) - 1  # perfect subtree
    return 1 + count_nodes(root.left) + count_nodes(root.right)

# Build 6-node complete tree
r = TreeNode(1,
    TreeNode(2, TreeNode(4), TreeNode(5)),
    TreeNode(3, TreeNode(6))
)
assert count_nodes(r) == 6
assert count_nodes(TreeNode(1)) == 1
assert count_nodes(None) == 0
print('All tests passed ✅')
````

## Hard 2 — Find Middle of Linked List (Fast-Slow Pointer)

Given a singly linked list, return the middle node. For even-length lists, return the second middle node.

**Example:**
```
Input:  1 -> 2 -> 3 -> 4 -> 5
Output: node with val 3

Input:  1 -> 2 -> 3 -> 4 -> 5 -> 6
Output: node with val 4
```

````python
def middle_node(head: ListNode) -> ListNode:
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
    return slow

assert middle_node(make_list([1,2,3,4,5])).val == 3
assert middle_node(make_list([1,2,3,4,5,6])).val == 4
assert middle_node(make_list([1])).val == 1
assert middle_node(make_list([1,2])).val == 2
print('All tests passed ✅')
````

## Hard 3 — Word Ladder (Shortest Transformation)
Find the shortest transformation from `beginWord` to `endWord` changing one letter at a time, each intermediate word in `wordList`. Return 0 if impossible.

**Example:**
```
Input:  beginWord="hit", endWord="cog",
        wordList=["hot","dot","dog","lot","log","cog"]
Output: 5

Input:  endWord not in wordList
Output: 0
```

````python
from collections import deque

def word_ladder(beginWord: str, endWord: str, wordList: list) -> int:
    word_set = set(wordList)
    if endWord not in word_set:
        return 0
    q = deque([(beginWord, 1)])
    visited = {beginWord}
    while q:
        word, steps = q.popleft()
        for i in range(len(word)):
            for ch in 'abcdefghijklmnopqrstuvwxyz':
                cand = word[:i] + ch + word[i+1:]
                if cand == endWord:
                    return steps + 1
                if cand in word_set and cand not in visited:
                    visited.add(cand)
                    q.append((cand, steps + 1))
    return 0

assert word_ladder('hit','cog',['hot','dot','dog','lot','log','cog']) == 5
assert word_ladder('hit','cog',['hot','dot','dog','lot','log']) == 0
print('All tests passed ✅')
````

## Hard 4 — Maximum Path Sum in Binary Tree

A path in a binary tree is any sequence of nodes where each pair of adjacent nodes has an edge. Find the path with the maximum sum (path need not go through root; can start and end at any node).

**Example:**
```
Tree:  -10
       /  \
      9   20
         /  \
        15    7
Output: 42   (path: 15 -> 20 -> 7)

Tree:  [2, -1]
Output: 2
```

````python
def max_path_sum(root: TreeNode) -> int:
    best = [float('-inf')]

    def dfs(node):
        if not node:
            return 0
        left  = max(dfs(node.left),  0)
        right = max(dfs(node.right), 0)
        best[0] = max(best[0], node.val + left + right)
        return node.val + max(left, right)

    dfs(root)
    return best[0]

t1 = TreeNode(-10, TreeNode(9), TreeNode(20, TreeNode(15), TreeNode(7)))
assert max_path_sum(t1) == 42
t2 = TreeNode(2, TreeNode(-1))
assert max_path_sum(t2) == 2
print('All tests passed ✅')
````

## Hard 5 — Match Parentheses with Wildcards

Given a string with `(`, `)`, and `*` (wildcard: can be `(`, `)`, or empty), return `True` if it can be a valid parenthesis string.

**Example:**
```
Input:  s = "(*)"
Output: True

Input:  s = "(*))"
Output: True

Input:  s = "(((*"
Output: False
```

````python
def check_valid_string(s: str) -> bool:
    # Track range of possible open-paren counts
    lo = hi = 0
    for ch in s:
        if ch == '(':
            lo += 1; hi += 1
        elif ch == ')':
            lo -= 1; hi -= 1
        else:  # '*'
            lo -= 1; hi += 1
        if hi < 0:
            return False
        lo = max(lo, 0)
    return lo == 0

assert check_valid_string("(*)") == True
assert check_valid_string("(*))") == True
assert check_valid_string("(((*") == False
print('All tests passed ✅')
````
