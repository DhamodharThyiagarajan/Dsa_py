### 1. The Foundation: Longest Common Subsequence (LCS)

This is where you introduce the **Top-Down Recursive Approach**. Because subsequences allow for gaps, we must explore multiple branching paths.

**Example**
* **Input:** `text1 = "abcde"`, `text2 = "ace"`
* **Output:** `3`

**The Recursive Mathematical State**
* If the characters match ($text1[i] == text2[j]$), the result is $1 + LCS(i+1, j+1)$.
* If they differ, the result is $\max(LCS(i+1, j), LCS(i, j+1))$.

#### Step A: Explain the Top-Down Code (Memoization)
Show them how we translate the raw recursive logic into code, and then add the `memo` dictionary (the "notepad") to eliminate redundant calculations.

```python
# ==================================================
# Top-Down Recursive Approach (Memoization)
# Longest Common Subsequence
# ==================================================

def longest_common_subsequence_top_down(text1, text2):
    # This is our "notepad" to store overlapping subproblems
    memo = {}

    def solve(i, j):
        # Base Case: If we hit the end of either string, no more matches exist
        if i == len(text1) or j == len(text2):
            return 0

        # Check the notepad: Have we solved this exact index pairing before?
        if (i, j) in memo:
            return memo[(i, j)]

        # Recursive Transition
        if text1[i] == text2[j]:
            # Match: Add 1 and move both pointers forward
            memo[(i, j)] = 1 + solve(i + 1, j + 1)
        else:
            # Mismatch: Branch out and find the maximum of skipping a char in text1 OR text2
            memo[(i, j)] = max(solve(i + 1, j), solve(i, j + 1))

        return memo[(i, j)]

    return solve(0, 0)


# ==================================================
# Test Case
# ==================================================
text1_test = "abcde"
text2_test = "ace"

print(f"Input: text1 = '{text1_test}', text2 = '{text2_test}'")
print("Output:", longest_common_subsequence_top_down(text1_test, text2_test))
# Expected Output: 3
```

## The "Pick or Skip" Recursive Approach (1D Array)

**The Concept**
To truly master Dynamic Programming, you must understand the "Pick or Skip" recursive tree. Let's look at a classic array problem: **House Robber**.

You are a thief planning to rob houses along a street. Each house has a certain amount of money stashed. The only rule stopping you from robbing every single house is that adjacent houses have connected security systems. If you rob two houses directly next to each other, the alarm will sound.

**The Logic**
At every single house, your brain makes a distinct choice:
1. **Pick it:** You steal the money from this house, but you are forced to skip the next house.
2. **Skip it:** You leave this house alone and move to the very next house to see your options.

You want to find the path that yields the maximum total money.



**Example**
* **Input:** `houses = [2, 7, 9, 3, 1]`
* **Output:** `12` (Rob house 1 for $2, house 3 for $9, and house 5 for $1).

```python
# ==================================================
# Top-Down Recursive Approach (Memoization)
# Maximum Sum of Non-Adjacent Elements (House Robber)
# ==================================================

def rob_houses(houses):
    # Our "notepad" to remember the maximum money we can make starting from house 'i'
    memo = {}

    # Define our recursive function
    def solve(i):
        # Base Case 1: If we have passed the last house, there is no more money to steal
        if i >= len(houses):
            return 0

        # Check the notepad: Have we calculated the max profit from this house onwards before?
        if i in memo:
            return memo[i]

        # The Core Decision: Pick vs. Skip

        # Option A (Pick): Steal from this house, plus whatever we can steal starting 2 houses down
        pick = houses[i] + solve(i + 2)

        # Option B (Skip): Ignore this house, and see what we can steal starting at the next house
        skip = solve(i + 1)

        # We want the absolute maximum of these two branching paths
        memo[i] = max(pick, skip)

        return memo[i]

    # Start the recursion at the very first house (index 0)
    return solve(0)


# ==================================================
# Test Cases
# ==================================================

print("Example 1:")
houses1 = [1, 2, 3, 1]
print(f"Input: {houses1}")
print("Output:", rob_houses(houses1))
# Expected Output: 4 (Pick 1, Skip 2, Pick 3, Skip 1)

print("\nExample 2:")
houses2 = [2, 7, 9, 3, 1]
print(f"Input: {houses2}")
print("Output:", rob_houses(houses2))
# Expected Output: 12 (Pick 2, Pick 9, Pick 1)
```


### Longest Common Substring

Use this to introduce the concept of comparing two strings character by character, but with a strict rule: the matching characters must be physically touching.

**Example**
* **Input:** `s1 = "ABCDGH"`, `s2 = "ACDGHR"`
* **Output:** `4` (The string is "CDGH")

**The Logic**
We iterate through both strings. If `s1[i]` matches `s2[j]`, the current streak is 1 plus whatever the streak was at the previous characters. If they do not match, the streak is completely broken, and the state drops to 0.

```python
def longest_common_substring_1d(s1, s2):
    m = len(s1)
    n = len(s2)

    # We only need a 1D array of size n + 1
    dp = [0] * (n + 1)
    max_length = 0

    # Iterate through characters of s1
    for i in range(1, m + 1):

        # CRITICAL: Iterate through s2 backwards!
        for j in range(n, 0, -1):

            # If characters match
            if s1[i - 1] == s2[j - 1]:
                dp[j] = dp[j - 1] + 1
                max_length = max(max_length, dp[j])

            # If they don't match
            else:
                dp[j] = 0

    return max_length

# Test Case
s1_test = "ABCDGH"
s2_test = "ACDGHR"
print("Output:", longest_common_substring_1d(s1_test, s2_test))
```

## The 1D Array Shift: Word Break

**Concept**
Not all string Dynamic Programming problems require comparing two different strings in a massive 2D matrix. Sometimes, we only need to evaluate a single string, allowing us to optimize our logic down to a 1-Dimensional array.

The **Word Break** problem asks: *Given a string and a dictionary of valid words, can the string be segmented into a space-separated sequence of dictionary words?*

**Example**
* **Input:** `s = "leetcode"`, `wordDict = ["leet", "code"]`
* **Output:** `True`

**The Logic**
We create a boolean array `dp` where `dp[i]` explicitly answers the question: *"Can the string strictly up to index `i` be formed by our dictionary?"* As we sweep through the string, we look backward. If the current chunk of letters is a valid dictionary word, AND the chunk immediately preceding it (`dp[i - length_of_word]`) was also marked as mathematically valid, we mark our current position as `True`.

**Time Complexity:** **O(N^2)** where N is the length of the string. The nested loops force us to evaluate the substring combinations, but dictionary lookups happen in constant time.

**Space Complexity:** **O(N)** to store our 1D boolean state array, plus the space required for the hash set of dictionary words.

```python
# ==================================================
# 1D Dynamic Programming Implementation
# Word Break Problem
# ==================================================

def word_break(s, wordDict):

    # Convert the word dictionary list into a set.
    # This optimizes our word lookups from O(N) to O(1) time complexity.
    word_set = set(wordDict)

    # Capture length of the string
    n = len(s)

    # Initialize a 1D DP boolean array.
    # dp[i] represents if the substring s[0...i-1] can be segmented.
    dp = [False] * (n + 1)

    # Base Case: An empty string can always be validly segmented.
    dp[0] = True

    # Outer loop: Sweep through the string to evaluate every end position 'i'
    for i in range(1, n + 1):

        # Inner loop: Check all possible previous split points 'j' before 'i'
        for j in range(i):

            # The core DP transition check:
            # 1. Is the string valid up to index 'j'? (dp[j])
            # 2. Is the remaining chunk from 'j' to 'i' a valid dictionary word?
            if dp[j] and s[j:i] in word_set:

                # If both are true, this point 'i' is officially a valid segment
                dp[i] = True

                # We found a valid split, no need to check further 'j' points for this 'i'
                break

    # The final answer rests at the very end of our boolean array
    return dp[n]


# ==================================================
# Test Cases
# ==================================================

print("Example 1:")
s1 = "leetcode"
dict1 = ["leet", "code"]
print(f"Input: '{s1}' | Dictionary: {dict1}")
print("Output:", word_break(s1, dict1))
# Expected Output: True

print("\nExample 2:")
s2 = "applepenapple"
dict2 = ["apple", "pen"]
print(f"Input: '{s2}' | Dictionary: {dict2}")
print("Output:", word_break(s2, dict2))
# Expected Output: True

print("\nExample 3:")
s3 = "catsandog"
dict3 = ["cats", "dog", "sand", "and", "cat"]
print(f"Input: '{s3}' | Dictionary: {dict3}")
print("Output:", word_break(s3, dict3))
# Expected Output: False
```

## Practice Questions

### 1. Reconstruct the Palindrome (Traceback)
* **Input:** `s = "bbbab"`
* **Output:** `"bbbb"`
* **Explanation:** The length of the longest palindromic subsequence is 4. By tracing backward through the populated DP matrix, matching outer boundary characters are appended to the left and right sides of the final result string, successfully reconstructing `"bbbb"`.

### 2. Weighted Edit Distance
* **Input:** `word1 = "horse"`, `word2 = "ros"` *(Assuming Insert/Delete cost 1, Replace costs 3)*
* **Output:** `4`
* **Explanation:** Because replacing `"h"` with `"r"` now costs 3 penalty points, the algorithm optimizes by deleting `"h"` (cost 1) and inserting `"r"` (cost 1). Dropping the `"e"` costs 1. Total cost = 3. This forces students to adapt standard transition equations to favor cheaper operational paths.

### 3. Is Subsequence
* **Input:** `s = "abc"`, `t = "ahbgdc"`
* **Output:** `True`
* **Explanation:** The characters `"a"`, `"b"`, and `"c"` are all found inside string `t` in the exact same chronological order. Students should realize this can be solved by proving the LCS of both strings is exactly mathematically equal to the length of `s`.

### 4. Delete Operation for Two Strings
* **Input:** `word1 = "sea"`, `word2 = "eat"`
* **Output:** `2`
* **Explanation:** You must delete `"s"` from `"sea"` to get `"ea"`, and delete `"t"` from `"eat"` to get `"ea"`. This takes exactly 2 steps. The core logic relies on finding the LCS (`"ea"`) and subtracting it from the total lengths of both original strings.

### 5. Minimum ASCII Delete Sum
* **Input:** `s1 = "sea"`, `s2 = "eat"`
* **Output:** `231`
* **Explanation:** Instead of counting the number of deletions, you sum the ASCII values of the deleted characters. Deleting `"s"` (ASCII 115) and `"t"` (ASCII 116) yields 115 + 116 = 231. Base cases and mismatch penalties must add `ord(char)` instead of a flat 1.

### 6. Distinct Subsequences
* **Input:** `s = "rabbbit"`, `t = "rabbit"`
* **Output:** `3`
* **Explanation:** There are exactly 3 distinct ways to generate the word `"rabbit"` from `"rabbbit"` by choosing which of the three `"b"`s to drop. When `s[i] == t[j]`, the DP state must combine the number of ways utilizing the match plus the number of ways skipping the match.

### 7. Interleaving String
* **Input:** `s1 = "aabcc"`, `s2 = "dbbca"`, `s3 = "aadbbcbcac"`
* **Output:** `True`
* **Explanation:** String `s3` is formed by perfectly weaving the characters of `s1` and `s2` together without violating the internal relative order of either original string. The DP grid must map boolean `True`/`False` states tracking valid pathways down and to the right.

### 8. Regular Expression Matching
* **Input:** `s = "aab"`, `p = "c*a*b"`
* **Output:** `True`
* **Explanation:** The `*` token strictly means "zero or more of the preceding character." Therefore, `c*` becomes zero `"c"`s, `a*` becomes two `"a"`s, and `b` matches the final `"b"`. The grid must handle looking back two columns to process the "zero occurrences" mathematical route.

### 9. Wildcard Matching
* **Input:** `s = "acdcb"`, `p = "a*c?b"`
* **Output:** `False`
* **Explanation:** The `?` matches any single character (it would match the `"d"`), and `*` matches any sequence. However, the sequence in the pattern requires a `"c"` immediately before the `?`, which fails against the input string. This challenges students to adapt regex logic into a simpler but broader wildcard ruleset.

### 10. Shortest Common Supersequence
* **Input:** `str1 = "abac"`, `str2 = "cab"`
* **Output:** `"cabac"`
* **Explanation:** `"cabac"` is the shortest possible string that contains both `"abac"` and `"cab"` as native subsequences. The algorithm requires computing the standard LCS first, and then performing a reverse traceback to print the non-matching characters combined with the shared LCS backbone.
