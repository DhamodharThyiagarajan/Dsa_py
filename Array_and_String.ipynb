## 1. Arrays (Lists in Python)
In Python, arrays are implemented as **Lists**. They are dynamic (they can grow/shrink) and **mutable** (you can change elements directly in memory).
* **Concept:** "Lists are Lockers." You can open locker #2 and swap the contents inside at any time without building a new row of lockers.

**Key Array Methods:**
* **Adding:** `arr.append(x)` (adds to end), `arr.insert(i, x)` (inserts at index), `arr.extend(list)` (merges lists).
* **Removing:** `arr.pop()` (removes last), `arr.remove(x)` (removes first occurrence).
* **Searching/Sorting:** `arr.index(x)`, `arr.count(x)`, `arr.sort()`, `arr.reverse()`.
* **Math/Utility:** `len(arr)`, `min(arr)`, `max(arr)`, `sum(arr)`.

## 2. Strings
Strings represent text. Unlike lists, strings in Python are **immutable**. You cannot change them once they are created.
* **Concept:** "Strings are Stones." Once carved, you cannot change a single letter on the stone. You must carve a completely new stone to make a change.

**Key String Methods:**
* **Case Conversion:** `s.upper()`, `s.lower()`, `s.title()`.
* **Searching:** `s.find(substring)`, `s.count(substring)`.
* **Checking:** `s.isalnum()`, `s.isdigit()`, `s.isalpha()`.
* **Modifying (Creates a new string!):** `s.replace(old, new)`, `s.strip()` (removes edge spaces), `s.split(" ")` (turns string into a list), `",".join(list)` (turns list into a string).

## 3. Common Operations for Both (Sequences)
* **Indexing:** `data[0]` gets the first element. `data[-1]` gets the last element.
* **Slicing:** `data[1:4]` gets elements from index 1 up to (but not including) 4.
* **Iteration:** `for item in data:` loops through elements one by one.


## Part 2: Pattern 1 - Two Pointers (Opposite Ends)

**The Concept:** When dealing with a sorted array or a string that needs to be evaluated from the outside in, initialize two variables: `left = 0` and `right = len(arr) - 1`. Move them towards each other until they meet.

**Why it's great:** It allows you to traverse data in $O(N)$ time while using $O(1)$ space.

### Question 1: Valid Palindrome (Core Example)
**Problem:** Check if a string reads the same forwards and backwards. Ignore case and non-alphanumeric characters.



```python
def is_palindrome(s: str):
    left = 0
    right = len(s) - 1

    while left < right:
        if not s[left].isalnum():
            left += 1
        elif not s[right].isalnum():
            right -= 1
        else:
            if s[left].lower() != s[right].lower():
                return False
            left += 1
            right -= 1
    return True

is_palindrome("hello world")
```

Question 2: Reverse a String In-PlaceProblem: Write a function that reverses a list of characters in-place (Space complexity must be $O(1)$).

```python
def reverse_string(s: list):
    # Initialize two pointers, one at the beginning and one at the end of the list.
    left = 0
    right = len(s) - 1

    # Continue swapping elements as long as the left pointer is before the right pointer.
    while left < right:
        # Swap the elements at the left and right pointers.
        s[left], s[right] = s[right], s[left]
        # Move the left pointer one step to the right.
        left += 1
        # Move the right pointer one step to the left.
        right -= 1
    # Return the modified list (in-place reversal).
    return s

reverse_string(["h", "e", "l", "l", "o"])
```

Question 3: Two Sum II (Sorted Array)
Problem: Given an array of integers sorted in non-decreasing order, find two numbers that add up to a specific target number. Return their indices.

```python
def two_sum_sorted(numbers: list, target: int):
    left = 0
    right = len(numbers) - 1

    while left < right:
        current_sum = numbers[left] + numbers[right]
        if current_sum == target:
            return [left, right]
        elif current_sum < target:
            left += 1 # We need a bigger number, move left pointer right
        else:
            right -= 1 # We need a smaller number, move right pointer left
    return []

two_sum_sorted([2, 7, 11, 15], 9)
```

Question 4: Sort Array by Parity
Problem: Given an integer array, move all the even integers to the beginning of the array followed by all the odd integers.

```python
def sort_array_by_parity(arr: list):
    left = 0
    right = len(arr) - 1

    while left < right:
        if arr[left] % 2 == 0:
            left += 1 # It's already even, move on
        elif arr[right] % 2 != 0:
            right -= 1 # It's already odd, move on
        else:
            # Left is odd, right is even. Swap them!
            arr[left], arr[right] = arr[right], arr[left]
            left += 1
            right -= 1
    return arr
sort_array_by_parity([3, 1, 2, 4])
```

 Pattern 2 - Frequency Counting (Hash Maps)The Concept: When a problem asks about finding duplicates, counting occurrences, or checking combinations, avoid using nested loops ($O(N^2)$). Instead, use a Python Dictionary to map items to their frequencies, reducing time complexity to $O(N)$.


Question 1: Valid Anagram (Core Example)Problem: Given two strings, return True if they are anagrams of each other (use the exact same letters in the exact same quantities).

```python
def is_anagram(s1: str, s2: str):
    if len(s1) != len(s2): return False

    freq = {}
    for char in s1:
        freq[char] = freq.get(char, 0) + 1

    for char in s2:
        if char not in freq or freq[char] == 0:
            return False
        freq[char] -= 1
    return True

is_anagram("anagram", "nagaram")
```

Question 2: First Unique Character in a String
Problem: Given a string, find the first non-repeating character and return its index. If it does not exist, return -1.

```python
def first_unique_char(s: str):
    freq = {}
    # First pass: Count frequencies
    for char in s:
        freq[char] = freq.get(char, 0) + 1

    # Second pass: Find the first one with a count of 1
    for index, char in enumerate(s):
        if freq[char] == 1:
            return index
    return -1

print(first_unique_char("leetcode"))
```

Question 3: Majority ElementProblem: Given an array of size $N$, find the majority element (the element that appears more than $N/2$ times).

```python
def majority_element(nums: list):
    freq = {}
    target_count = len(nums) // 2

    for num in nums:
        freq[num] = freq.get(num, 0) + 1
        # Check immediately if it hit the threshold
        if freq[num] > target_count:
            return num

majority_element([2, 2, 1, 1, 1, 2, 2])
```

Question 4: Contains Duplicate
Problem: Given an integer array, return True if any value appears at least twice in the array, and return False if every element is distinct.

```python
def contains_duplicate(nums: list):
    seen = {} # Can also use a Python set() here!

    for num in nums:
        if num in seen:
            return True
        seen[num] = True
    return False
contains_duplicate([1, 2, 3, 1])
```

Question 5: Single Number
Problem: Given a non-empty array of integers, every element appears twice except for one. Find that single one.

```python
def single_number(nums: list):
    freq = {}

    for num in nums:
        freq[num] = freq.get(num, 0) + 1

    for num, count in freq.items():
        if count == 1:
            return num
single_number([2, 2, 1])
```

> **Teacher's Hint:** Start like a normal valid palindrome using two pointers. If you hit a mismatch, you have two choices: skip the left character OR skip the right character. If either resulting substring is a palindrome, return `True`!

### 4. Count Pairs Whose Sum is Less than Target
**Problem:** Given an array of integers that is already sorted, and a `target` integer, return the number of pairs `(i, j)` where `i < j` and `nums[i] + nums[j] < target`.

* **Output:** `3`
> **Teacher's Hint:** If `nums[left] + nums[right] < target`, it means `nums[left]` paired with any number between `left` and `right` will also be less than the target! Add `(right - left)` to your total count and move the left pointer up.

### 5. Container With Most Water

* **Output:** `49`
> **Teacher's Hint:** The width is `right - left`. The height is the minimum of `heights[left]` and `heights[right]`. Calculate the area. To find a potentially bigger area, always move the pointer that points to the shorter line inward.
