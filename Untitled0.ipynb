1) Bubble_sort

```python
def bubble_sort(arr):
  n=len(arr)
  for i in range(n):
    swapped=False
    for j in range(0,n-i-1):
      if arr[j]>arr[j+1]:
        arr[j],arr[j+1]=arr[j+1],arr[j]
        swapped=True
    if not swapped:
      break

  return arr

print("Bubble Sort:", bubble_sort([64, 34, 25, 12, 22, 11, 90]))
```

1.Move Zeroes to the End

Problem: Given an integer array, move all 0s to the end while maintaining the relative order of the non-zero elements. Approach: This uses modified Bubble Sort logic. Instead of bubbling the absolute largest element, you bubble the 0s to the right by swapping them with adjacent non-zero elements.

```python
def move_zeroes(arr):
  last_non_zero_inded=0
  for i in range(len(arr)):
    if arr[i]!=0:
      arr[last_non_zero_inded],arr[i]=arr[i],arr[last_non_zero_inded]
      last_non_zero_inded+=1
  return arr

print("Move Zeroes:", move_zeroes([0, 1, 0, 3, 12]))
```

2. Sort an Array in Descending Order

Problem: Modify standard Bubble Sort to sort elements from highest to lowest.
Approach: Simply flip the comparison operator. If arr[j] < arr[j + 1], swap them. This forces the smallest elements to "bubble" to the end.

```python
def descending_order(arr):
  n=len(arr)
  for i in range(n):
    swapped=False
    for j in range(0,n-i-1):
      if arr[j]<arr[j+1]:
        arr[j],arr[j+1]=arr[j+1],arr[j]
        swapped=True
    if not swapped:
      break
  return arr

print(descending_order([5, 1, 4, 2, 8]))
```

3. Sort an Array of Strings by Length

Problem: Given an array of strings, sort them so that shorter strings appear first.
Approach: Use the Bubble Sort iteration, but compare the len(arr[j]) and len(arr[j+1]) instead of the alphabetical characters.

```python
def sort_array_of_strings_by_length(arr):
  n=len(arr)
  for i in range(n):
    swapped=False
    for j in range (0,n-i-1):
      if len(arr[j])> len(arr[j+1]):
        arr[j],arr[j+1]=arr[j+1],arr[j]
        swapped=True
    if not swapped:
      break
  return arr

words = ["apple", "kiwi", "banana", "fig", "grapefruit"]
print(sort_array_of_strings_by_length(words))
```

4. Check if an Array is Sorted in One Pass

Problem: Write a function that returns True if an array is sorted, using the mechanics of Bubble Sort.
Approach: Utilize the swapped flag optimization. Run exactly one pass of the inner loop. If no swaps occur, return True. This executes in  O(n)  time.

```python
def is_sorted(arr):
  n=len(arr)
  swapped=False

  for j in range(n-1):
    if arr[j]>arr[j+1]:
      arr[j], arr[j + 1] = arr[j + 1], arr[j]
      swapped = True

  return not swapped


print(is_sorted([1, 2, 3, 4, 5]))
print(is_sorted([5, 3, 4, 1, 2]))
```

5. Sort by Absolute Difference from a Target Value X

Problem: Given an array and a value X, sort the array based on the absolute difference between the elements and X.
Approach: During the adjacent element comparison step, evaluate abs(arr[j] - X) > abs(arr[j+1] - X). If true, swap the elements.

```python
def sort_by_difference_from_target(arr,x):
  n=len(arr)
  for i in range(n):
    swapped=False
    for j in range(0,n-i-1):
      if abs(arr[j]-x) > abs(arr[j+1]-x):
        arr[j], arr[j + 1] = arr[j + 1], arr[j]
        swapped = True
    if not swapped:
      break
  return arr

nums = [10, 5, 3, 9, 2]
X = 7
print(sort_by_difference_from_target(nums, X))
# Output: [5, 9, 10, 3, 2]
# 9 (2), 5 (2), 10 (3), 3 (4), 2 (5)]
```

2.selection sort

```python
def selection_sort(arr):
  n=len(arr)
  for i in range(n-1):
    min_index=i
    for j in range(i+1,n):
      if arr[j]<arr[min_index]:
        min_index=j
    if arr[i]>arr[min_index]:   #OR min_index!=i:
      arr[i],arr[min_index]=arr[min_index],arr[i]
  return arr

print("Selection Sort:", selection_sort([29, 10, 14, 37, 13]))
```

1.Find the Kth Smallest Element

Problem: Find the Kth smallest element in an unsorted array without fully sorting it. Approach: Run the outer loop of Selection Sort exactly K times. After K passes, the Kth smallest element will be locked into the arr[K-1] position.




```python
def kth_smallest(arr,k):
  n=len(arr)
  for i in range(k):
    min_index=i
    for j in range(i+1,n):
      if arr[j]<arr[min_index]:
        min_index=j
    if arr[i]>arr[min_index]:
      arr[i],arr[min_index]=arr[min_index],arr[i]
  return arr[k-1]
print("3rd Smallest Element:", kth_smallest([7, 10, 4, 3, 20, 15], 3))
# Expected: 7 (The sorted array would be [3, 4, 7, 10, 15, 20])
```

2. Stable Selection Sort

Problem: Standard Selection Sort is unstable. Modify it so that identical elements maintain their relative initial order.
Approach: Instead of swapping the found minimum element with the first unsorted element, shift all elements from the first unsorted index up to the minimum element's index one position to the right, then place the minimum element in the freed space.

```python
def stable_selection_sort(arr):
    n = len(arr)
    for i in range(n - 1):
        # Find the minimum element in the unsorted part
        min_index = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_index]:
                min_index = j

        # Instead of swapping, we shift elements to the right
        key = arr[min_index]
        while min_index > i:
            arr[min_index] = arr[min_index - 1]
            min_index -= 1
        arr[i] = key

    return arr

# Example
nums = [4, 5, 3, 2, 4, 1]
print(stable_selection_sort(nums))
# Output: [1, 2, 3, 4, 4, 5]
```

 3.Sort a Linked List using Selection Sort

Problem: Apply the Selection Sort algorithm to a singly linked list.
Approach: Traverse the linked list with a pointer. For each node, find the minimum node in the remaining list and swap their values.

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

def selection_sort_linked_list(head):
    # Traverse the linked list
    current = head
    while current:
        # Assume current node is the minimum
        min_node = current
        next_node = current.next

        # Find the minimum node in the remaining list
        while next_node:
            if next_node.data < min_node.data:
                min_node = next_node
            next_node = next_node.next

        # Swap values (not nodes)
        current.data, min_node.data = min_node.data, current.data

        # Move to the next node
        current = current.next

    return head

# Helper function to print the list
def print_list(head):
    temp = head
    while temp:
        print(temp.data, end=" -> ")
        temp = temp.next
    print("None")

# Example usage
head = Node(4)
head.next = Node(2)
head.next.next = Node(1)
head.next.next.next = Node(3)

print("Original List:")
print_list(head)

sorted_head = selection_sort_linked_list(head)

print("Sorted List:")
print_list(sorted_head)
```

4. Maximize Array Sum After K Negations

Problem: Given an array, negate exactly K elements to maximize the total sum of the array.
Approach: Use a Selection Sort mindset: iteratively find the absolute minimum value in the array K times and multiply it by -1.

```python
def maximize_sum(arr,k):
  n=len(arr)
  for _ in range(k):
    min_index=0
    for i in range(1,n):
      if arr[i]< arr[min_index]:
        min_index=i
    arr[min_index]=-arr[min_index]
  return sum(arr)
nums = [-2, 0, 5, -1, 2]
K = 4
print(maximize_sum(nums, K))  # Output: 10
```

5. Sort Half Ascending, Half Descending

Problem: Sort the first half of an array in ascending order and the second half in descending order.
Approach: Run standard Selection Sort on the first half (finding minimums). Run a modified Selection Sort on the second half (finding maximums).

```python
def sort_half(arr):
    n = len(arr)
    mid = n // 2

    # First half: Ascending Selection Sort
    for i in range(mid):
        min_index = i
        for j in range(i+1, mid):
            if arr[j] < arr[min_index]:
                min_index = j
        arr[i], arr[min_index] = arr[min_index], arr[i]

    # Second half: Descending Selection Sort
    for i in range(mid, n):
        max_index = i
        for j in range(i+1, n):
            if arr[j] > arr[max_index]:
                max_index = j
        arr[i], arr[max_index] = arr[max_index], arr[i]

    return arr

# Example
nums = [7, 2, 9, 4, 1, 5, 8, 3]
print(sort_half(nums))  # Output: [2, 4, 7, 9, 8, 5, 3, 1]
```

3.insertion sort


```python
def insertion_sort(arr):
  n=len(arr)
  for i in range(1,n):
    key=arr[i]
    j=i-1
    while j>=0 and key<arr[j]:
      arr[j+1]=arr[j]
      j-=1
    arr[j+1]=key
  return arr
print("Insertion Sort:", insertion_sort([12, 11, 13, 5, 6]))
```

1. Binary Insertion Sort

Problem: Improve the standard Insertion Sort algorithm to reduce the number of comparisons.
Approach: Instead of linearly searching backward to find the insertion point for the key, use Binary Search on the already sorted portion of the array to find the correct index in  O(logn)  comparisons.

```python
def binary_search(arr, key, start, end):
    while start <= end:
        mid = (start + end) // 2
        if arr[mid] < key:
            start = mid + 1
        else:
            end = mid - 1
    return start

def binary_insertion_sort(arr):
  n=len(arr)
  for i in range(1,n):
    key=arr[i]
    # Find insertion position using binary search
    pos=binary_search(arr,key,0,i-1)
    # Shift elements to the right
    arr[pos+1:i+1]=arr[pos:i]
    arr[pos] = key
  return arr

nums = [37, 23, 0, 17, 12, 72, 31]
print("Sorted:", binary_insertion_sort(nums))
```

3. Insertion Sort on a Singly Linked List

Problem: Implement Insertion Sort for a linked list where you can only traverse forward.
Approach: Create a dummy node representing the sorted list. Iterate through the original list, and for each node, traverse the new sorted list from the head to find its exact insertion point, updating pointers accordingly.

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

def insertion_sort_linked_list(head):
    # Dummy node as the new sorted list head
    dummy = Node(0)
    current = head

    while current:
        # At each iteration, insert current into sorted list
        prev = dummy
        next_node = current.next

        # Find insertion point
        while prev.next and prev.next.data < current.data:
            prev = prev.next

        # Insert current between prev and prev.next
        current.next = prev.next  #dummy -> 3 (c)-> 4 -> None
        prev.next = current

        # Move to next node in original list
        current = next_node

    return dummy.next

# Helper function to print list
def print_list(head):
    temp = head
    while temp:
        print(temp.data, end=" -> ")
        temp = temp.next
    print("None")

# Example usage
head = Node(4)
head.next = Node(3)
head.next.next = Node(1)
head.next.next.next = Node(2)

print("Original List:")
print_list(head)

sorted_head = insertion_sort_linked_list(head)

print("Sorted List:")
print_list(sorted_head)
```

5. Sort an Array of Objects by Attribute

Problem: Given an array of Employee dictionaries, sort them by age in ascending order.
Approach: Modify the comparison step in the while loop to explicitly evaluate key['age'] < arr[j]['age'].

```python
def insertion_sort_by_age(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        # Compare based on 'age'
        while j >= 0 and key['age'] < arr[j]['age']:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr

# Example usage
employees = [
    {"name": "Alice", "age": 34},
    {"name": "Bob", "age": 28},
    {"name": "Charlie", "age": 40},
    {"name": "David", "age": 25}
]

sorted_employees = insertion_sort_by_age(employees)
print(sorted_employees)
```
