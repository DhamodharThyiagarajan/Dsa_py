# Dynamic Programming

Dynamic Programming (DP) is a method for solving complex problems by breaking them down into simpler, overlapping subproblems. Instead of recalculating the same subproblems (brute force), we store their results in an array or matrix and reuse them.

Always remember the **S.T.B.R.** framework:
* **State:** What does one cell in our memory array represent?
* **Transition:** What is the formula linking the past to the present?
* **Base Case:** How do we initialize the start to prevent errors?
* **Result:** Where is the final answer stored?



# Introduction to Dynamic Programming: The Fibonacci Sequence

Before we dive into arrays and matrices, let's look at the most famous Dynamic Programming problem in the world: the Fibonacci sequence.

In a Fibonacci sequence, every number is the sum of the two preceding ones: 0, 1, 1, 2, 3, 5, 8, 13...

If we try to calculate the 5th Fibonacci number recursively, the computer calculates $F(3)$ multiple times, $F(2)$ multiple times, and so on. It forms a massive, redundant tree of operations.

**Dynamic Programming fixes this by remembering the past.** Instead of recalculating, we write down the answer for $F(2)$ the first time we see it. When we need it again, we just look it up.

Let's apply our **S.T.B.R.** framework:
* **State:** $dp[i]$ represents the $i$-th number in the Fibonacci sequence.
* **Transition:** $dp[i] = dp[i-1] + dp[i-2]$ (The core math rule).
* **Base Case:** $dp[0] = 0$ and $dp[1] = 1$ (The starting numbers).
* **Result:** The final answer is stored at the very end of our array at $dp[n]$.

```python
import time

# ==========================================
# 1. THE BAD WAY: Naive Recursion
# ==========================================
def fib_recursive(n):
    # Base cases
    if n == 0: return 0
    if n == 1: return 1

    # Transition (Recalculating over and over)
    return fib_recursive(n - 1) + fib_recursive(n - 2)

# ==========================================
# 2. THE DP WAY: Bottom-Up Tabulation
# ==========================================
def fib_dp(n):
    # Edge case
    if n == 0: return 0
    if n == 1: return 1

    # State: Create our memory array
    dp = [0] * (n + 1)

    # Base Cases
    dp[0] = 0
    dp[1] = 1

    # Transition: Build from the bottom up
    for i in range(2, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]

    # Result
    return dp[n]

# ==========================================
# Let's see the difference in action
# ==========================================
target_number = 35 # High enough to make recursion struggle

print(f"Calculating Fibonacci({target_number})...\n")

# Test DP
start_dp = time.time()
result_dp = fib_dp(target_number)
end_dp = time.time()
print(f"DP Result: {result_dp}")
print(f"DP Time: {end_dp - start_dp:.6f} seconds (Instant)\n")

# Test Recursion
start_rec = time.time()
result_rec = fib_recursive(target_number)
end_rec = time.time()
print(f"Recursion Result: {result_rec}")
print(f"Recursion Time: {end_rec - start_rec:.6f} seconds (Slow!)")
```

## Question 1: Longest Increasing Subsequence
The following code evaluates the historical subset to find the longest strictly increasing sequence.

```python
def length_of_lis(nums):
    # Base case: an empty array has a subsequence of length 0
    if not nums:
        return 0

    # Initialize a DP array of the same length as nums, filled with 1s.
    # Every single element is inherently an increasing subsequence of length 1.
    dp = [1] * len(nums)

    # Outer loop: The current element we are evaluating
    for i in range(len(nums)):
        # Inner loop: Look back at all previous elements
        for j in range(i):
            # If the current element is larger than the previous element,
            # we can mathematically extend that previous subsequence.
            if nums[i] > nums[j]:
                dp[i] = max(dp[i], dp[j] + 1)

    # The maximum value in the DP array is our final answer
    return max(dp)

# Example Execution
example_input = [10, 9, 2, 5, 3, 7, 101, 18]
print("Input:", example_input)
print("Longest Increasing Subsequence Length:", length_of_lis(example_input))
```

## Question 2: Maximum Subarray Sum
**The Concept:** Kadane's Algorithm. At each step, we decide whether to add the current number to our running sum, or if the running sum is so negative that it's better to just start a brand new sequence at the current number.

```python
def max_sub_array(nums):
    if not nums: return 0

    current_max = global_max = nums[0]

    for i in range(1, len(nums)):
        # Transition: Start fresh at nums[i], or add nums[i] to previous sum
        current_max = max(nums[i], current_max + nums[i])
        # Update global record
        global_max = max(global_max, current_max)

    return global_max

print("Maximum Subarray Sum:", max_sub_array([-2, 1, -3, 4, -1, 2, 1, -5, 4]))
```


## Question 3: Subset Sum
**The Concept:** We use a 1D boolean array where each index represents a possible target sum. We iterate backward to prevent using the same number multiple times.

```python
def can_partition_subset_sum(nums, target):
    # DP array stores True if a specific sum is achievable
    dp = [False] * (target + 1)
    dp[0] = True # Base case: target 0 is always achievable (empty subset)

    for num in nums:
        # Traverse backwards to prevent infinite reuse of the same element
        for current_target in range(target, num - 1, -1):
            dp[current_target] = dp[current_target] or dp[current_target - num]

    return dp[target]

print("Subset Sum Possible:", can_partition_subset_sum([3, 34, 4, 12, 5, 2], 9))
```
