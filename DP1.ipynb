```python
import time

def fib(n):
  if n<=0:
    return 0
  elif n<=1:
    return 1
  else:
    return fib(n-1)+fib(n-2)
start=time.time()
print(fib(40))
end=time.time()
print(f'time required {end-start:.6f}')
```

```python
def fib_db(n):
  if n==0:return 0
  if n==1:return 1

  dp=[0]*(n+1)

  dp[0]=0
  dp[1]=1

  for i in range(2,n+1):
    dp[i]=dp[i-1]+dp[i-2]
  return dp[n]

start=time.time()
print(fib_db(40))
end=time.time()
print(f'time required {end-start:.6f}')
```

```python
def length_of_lis(nums):
  if not nums:
    return 0
  dp=[1]*len(nums)
  for i in range(len(nums)):
    for j in range(i):
      if nums[i]>nums[j]:
        dp[i]=max(dp[i],dp[j]+1)
  return max(dp)
print(length_of_lis([10,9,2,5,3,7,101,18]))
```

```python
def max_sub_arraysum(nums):
  if not nums:
    return 0
  current_sum=global_sum=nums[0]
  for i in range(1,len(nums)):
    current_sum=max(nums[i],current_sum+nums[i])
    global_sum=max(global_sum,current_sum)
  return global_sum
print(max_sub_arraysum([-2,1,-3,4,-1,2,1,-5,4]))
```
