## Medium 1 — Partition String Into Minimum Segments

Partition a string so each character appears in at most one segment. Return list of segment lengths (greedy: each segment extends to the last occurrence of every character inside it).

**Example:**
```
Input:  s = "ababcbacadefegdehijhklij"
Output: [9, 7, 8]

Input:  s = "eccbbbbdec"
Output: [10]
```

````python
def partition_string(s):
  last={c:i for i,c in enumerate(s)}
  start=end=0
  result=[]
  for i,c in enumerate(s):
    end=max(end,last[c])
    if i==end:
      result.append(end-start+1)
      start=i+1
  return result


print(partition_string("ababcbacadefegdehijhklij"))
````

## Medium 2 — Count Subarrays with Product Below Threshold

Given a positive integer array and threshold `k`, return count of contiguous subarrays whose product is strictly less than `k`.

**Example:**
```
Input:  nums = [10,5,2,6], k = 100
Output: 8
```

````python
def num_subarray_product_less_than_k(nums, k):
  if k<=1:
    return 0
  count=0
  product=1
  left=0
  for right in range(len(nums)):
    product *= nums[right]
    while product >= k : # only lesser than so
      product //= nums[left]
      left+=1
    count+=right-left+1
  return count
print(num_subarray_product_less_than_k([10,5,2,6], 100) )
````

```

Output: True

Input:  nums = [2,1,3,4]
Output: False
```

## Medium 4 — Zero Out Rows and Columns

Given an m×n matrix, if any cell is 0, set its entire row and column to 0 in-place.

**Example:**
```
Input:  [[1,1,1],[1,0,1],[1,1,1]]
Output: [[1,0,1],[0,0,0],[1,0,1]]
```

````python
def set_zeroes(matrix):
  row,col=set(),set()
  n=len(matrix)
  m=len(matrix[0])
  for i in range(n):
    for j in range(m):
      if matrix[i][j]==0:
        row.add(i)
        col.add(j)
  for i in range(n):
    for j in range(m):
      if i in row or j in col:
        matrix[i][j]=0
  return matrix
print(set_zeroes([[1,1,1],[1,0,1],[1,1,1]]))
````

```
Input:  s = "tree"

Input:  s = "cccaaa"
Output: "cccaaa"  (or "aaaccc")
```

````python
from collections import Counter
def frequency_sort(s):
  freq=Counter(s)
  sorted_chars=sorted(freq.items(),key=lambda x: -x[1])

  result= ''.join(char * count for char,count in sorted_chars)
  return result

print(frequency_sort("tree"))
````

## Medium 6 — Unique Paths in Grid with Obstacles

An m×n grid has obstacles (`1`). Count unique paths from top-left to bottom-right moving only right or down.

**Example:**
```
Input:  obstacleGrid = [[0,0,0],[0,1,0],[0,0,0]]
Output: 2

Input:  obstacleGrid = [[0,1],[0,0]]
Output: 1
```

````python
def unique_paths_with_obstacles(grid):
  m,n=len(grid),len(grid[0])
  if grid[0][0]==1 or grid[m-1][n-1]==1:
    return 0 #ways
  dp=[[0]*n for _ in range(m)]
  dp[0][0]=1 #ways
  for i in range(m):
    for j in range(n):
      if i==0 and j==0:
        continue
      if grid[i][j]==1: #If current obstacle 1 ,0 ways
        dp[i][j]=0
      else:
        dp[i][j]= (dp[i-1][j] if i > 0 else 0) + (dp[i][j-1] if j > 0 else 0) #top + left
  return dp[m-1][n-1]
print(unique_paths_with_obstacles([[0,0,0],[0,1,0],[0,0,0]]))
````

## Medium 8 — Validate Binary Search Tree

Return `True` if the binary tree is a valid BST (left subtree < node < right subtree, strictly, for all nodes).

**Example:**
```

    5

     / \

```

## Medium 10 — Remove Duplicates from Sorted Linked List

Given a sorted linked list, delete all nodes with duplicate values, leaving only distinct values.

**Example:**
```
Input:  1 -> 1 -> 2 -> 3 -> 3
Output: 1 -> 2 -> 3

Input:  1 -> 1 -> 1
Output: 1
```

````python
class ListNode:
  def __init__(self,val):
    self.val=val
    self.next=None

def make_list(vals):
    dummy=ListNode(0)
    temp=dummy
    for v in vals:
      temp.next=ListNode(v)
      temp=temp.next
    return dummy.next

def delete_duplicates(head):
    temp=head
    while temp and temp.next:
      if temp.val==temp.next.val:
        temp.next=temp.next.next
      else:
        temp=temp.next
    return head

def to_list(head):
    out = []
    while head:
        out.append(head.val)
        head = head.next
    return out

print(to_list(delete_duplicates(make_list([1,1,2,3,3]))))

 # head = make_list([1,1,2,3,3])   # head points to the first node (value 1)
 # head = delete_duplicates(head)  # head still points to 1, but list is now 1→2→3
 # print(to_list(head))            # traverses from head, returns [1,2,3]
````

## Medium 11 — Check Palindrome Linked List

Return `True` if the linked list is a palindrome. Solve in O(n) time and O(1) space.

**Example:**
```
Input:  1 -> 2 -> 2 -> 1
Output: True

Input:  1 -> 2
Output: False
```

````python
class ListNode:
  def __init__(self,val):
    self.val=val
    self.next=None

def make_list(vals):
    dummy=ListNode(0)
    temp=dummy
    for v in vals:
      temp.next=ListNode(v)
      temp=temp.next
    return dummy.next

def palindrome_linked_list(head):
  slow=fast=head
  while fast and fast.next:
    slow=slow.next
    fast=fast.next.next

  prev=None
  while slow:
    next_node=slow.next
    slow.next=prev
    prev=slow
    slow=next_node

  left=head
  right=prev
  while right:
    if left.val!=right.val:
      return False
    left=left.next
    right=right.next
  return True

print(palindrome_linked_list(make_list([1,2,2,1])))
````

## Medium 13 — Evaluate Postfix (RPN) Expression

Evaluate a Reverse Polish Notation expression. Operators: `+`, `-`, `*`, `/` (integer division truncated toward zero).

**Example:**
```
Input:  tokens = ["2","1","+","3","*"]
Output: 9

Input:  tokens = ["4","13","5","/","+"]
Output: 6
```

````python
import operator
def eval_tokens(tokens):
  ops={'+': operator.add, '-': operator.sub, '*': operator.mul,'/': lambda a, b: int(a / b)}
  stack=[]
  for t in tokens:
    if t in ops:
      b,a=stack.pop(),stack.pop()
      stack.append(ops[t](a,b))
    else:
      stack.append(int(t))
  return stack[0]

print( eval_tokens(["2","1","+","3","*"]))
````

## Medium 14 — Validate Bracket String

Given a string of `(`, `)`, `{`, `}`, `[`, `]`, return `True` if it is correctly matched and nested.

**Example:**
```
Input:  s = "()[]{}"
Output: True

Input:  s = "(]"
Output: False

Input:  s = "{[]}"
Output: True
```

````python
def valid_string(s):
  match = {')':'(', ']':'[', '}':'{'}
  stack=[]
  for ch in s:
    if ch in match:
      if not stack or stack[-1] !=match[ch]:
        return False
      stack.pop()
    else:
      stack.append(ch)
  return not stack

print(valid_string("()[]{}"))
````

## Medium 15 — Jump Game
Given an array where each element is the max jump from that index, return `True` if you can reach the last index.

**Example:**
```
Input:  nums = [2,3,1,1,4]
Output: True

Input:  nums = [3,2,1,0,4]
Output: False
```

````python
def jump_game(nums):
  reach=0
  for i,v in enumerate(nums):
    if i>reach:
      return False
    reach=max(reach,i+v)
  return True

print(jump_game([2,3,1,1,4]))
````

## Medium 16 — Kth Largest Element in an Array
Find the k-th largest element using a min-heap of size k.

**Example:**
```
Input:  nums = [3,2,1,5,6,4], k = 2
Output: 5

Input:  nums = [3,2,3,1,2,4,5,5,6], k = 4
Output: 4
```

````python
import heapq
def find_kth_largesr(nums,k):
  heap=[]
  for n in nums:
    heapq.heappush(heap,n)# push element into min‑heap
    if len(heap)>k:
      heapq.heappop(heap)
  return heap[0]

print(find_kth_largesr([3,2,1,5,6,4],2))
````

## Medium 17 — Longest Consecutive Sequence
Find the length of the longest consecutive elements sequence in O(n).

**Example:**
```
Input:  nums = [100,4,200,1,3,2]
Output: 4

Input:  nums = [0,3,7,2,5,8,4,6,0,1]
Output: 9
```

````python
def longest_consecutive_sequence(nums):
  num_set=set(nums)
  longest=0

  for n in num_set:
    if n-1 not in num_set:
      length=1
      while n + length in num_set:
        length+=1
      longest=max(longest,length)
  return longest

print(longest_consecutive_sequence([100,4,200,1,3,2]))

````

## Medium 19 — Rearrange Deck for Ascending Reveal Order

Arrange a deck so that repeatedly revealing the top card and moving the next to the bottom reveals cards in ascending order.

**Example:**
```
Input:  cards = [17,13,11,2,3,5,7]
Output: [2,13,3,11,5,17,7]

````python
from collections import deque

def reveal_increasing(deck):
  n=len(deck)
  deck.sort() #Sorted cards: [2,3,5,7,11,13,17]
  indices=deque(range(n)) #Indices queue: [0,1,2,3,4,5,6]
  result=[0]*n
  for card in deck:
    result[indices.popleft()]=card #[2,,,,,,]
    if indices:
      indices.append(indices.popleft()) #[2,3,4,5,6,1]
  return result


print(reveal_increasing([17,13,11,2,3,5,7]))
````

## Medium 20 — Car Fleet Count

Cars on a road each have a position and speed. Cars that catch up to a slower car ahead form a fleet. Return the number of fleets reaching the target.

**Example:**
```
Input:  target=12, position=[10,8,0,5,3], speed=[2,4,1,1,3]
Output: 3
```

````python
def car_fleet(target,postion,speed):
  stack=[]
  for pos,spd in sorted(zip(postion,speed),reverse=True): #[(10,2), (8,4), (5,1), (3,3), (0,1)]
    time=(target-pos)/spd
    if not stack or time>stack[-1]:
      stack.append(time)
  return len(stack)

print(car_fleet(12,[10,8,0,5,3],[2,4,1,1,3]))
````

## Medium 21 — Minimum Intervals to Complete All Tasks

Given a list of tasks (letters) and cooldown `n`, find the minimum CPU intervals needed. Idle slots are inserted when a task must wait.

**Example:**
```
Input:  tasks = ['A','A','A','B','B','B'], n = 2
Output: 8

Input:  tasks = ['A','A','A','B','B','B'], n = 0
Output: 6
```

````python
from collections import Counter
def least_intervals(tasks,n):
  freq=Counter(tasks)                    # A=3, B=3 → max_freq=3, max_count=2
  max_freq=max(freq.values())
  max_count=sum(1 for v in freq.values() if v==max_freq)
  return max(len(tasks),(max_freq-1)*(n+1)+max_count)

print(least_intervals(['A','A','A','B','B','B'],2))
````

## Hard 1 — Count All Nodes in Complete Binary Tree

Given a complete binary tree, count its nodes in O(log²n) time by exploiting the complete property (compare left and right heights).

**Example:**
```
Tree (complete, 6 nodes):
        1
       / \
      2   3
     / \ /
    4  5 6
Output: 6
```

````python
class TreeNode:
  def __init__(self,val):
    self.val=val
    self.left=None
    self.right=None

def build_tree(values):
  if not values or values[0]==None:
    return None
  root=TreeNode((values[0])) #5
  queue=deque([root]) #[5]
  i=1
  while queue and i<len(values):
    node=queue.popleft()#5
    if values[i]is not None:
      node.left=TreeNode((values[i]))
      queue.append(node.left)
    i+=1
    if i<len(values) and values[i]is not None:
      node.right=TreeNode((values[i]))
      queue.append(node.right)
    i+=1
  return root

def count_nodes(root):
    if not root:
      return 0
    def left_height(node):
      h=0
      while node:
        h+=1
        node=node.left
      return h
    def right_nodes(node):
      h=0
      while node:
        h+=1
        node=node.right
      return h

    lh,rh=left_height(root),right_nodes(root)

    if lh==rh:
      return (1 << lh)-1 #2^k -1 #perfect subtree
    else:
      return 1+count_nodes(root.left) + count_nodes(root.right)


print(count_nodes(build_tree([1,2,3,4,5,6])))
````

## Hard 3 — Word Ladder (Shortest Transformation)
Find the shortest transformation from `beginWord` to `endWord` changing one letter at a time, each intermediate word in `wordList`. Return 0 if impossible.

**Example:**
```
Input:  beginWord="hit", endWord="cog",
        wordList=["hot","dot","dog","lot","log","cog"]
Output: 5

Input:  endWord not in wordList
Output: 0
```

````python
from collections import deque
def word_ladder(beginword,endword,wordList):
  wordset=set(wordList)
  if endword not in wordset:
    return 0
  q=deque([(beginword,1)])
  visited={beginword}
  while q:
    word,steps=q.popleft()
    for i in range(len(word)):
      for ch in 'abcdefghijklmnopqrstuvwxyz':
        cand=word[:i]+ch+word[i+1:] # h i t
        if cand==endword:
          return steps+1
        if cand in wordset and cand not in visited:
          visited.add(cand)
          q.append((cand,steps+1))
  return 0

print(word_ladder("hit","cog",["hot","dot","dog","lot","log","cog"]))
````

## Hard 4 — Maximum Path Sum in Binary Tree

A path in a binary tree is any sequence of nodes where each pair of adjacent nodes has an edge. Find the path with the maximum sum (path need not go through root; can start and end at any node).

**Example:**
```
Tree:  -10
       /  \
      9   20
         /  \
        15    7
Output: 42   (path: 15 -> 20 -> 7)

Tree:  [2, -1]
Output: 2
```

````python
from collections import deque
class TreeNode:
  def __init__(self,val):
    self.val=val
    self.left=None
    self.right=None

def max_path_sum(root):
    best = [float('-inf')]
    def get_max_gain(node):
        if not node:
            return 0
        left_gain  = max(get_max_gain(node.left),  0) # if negative we drop so 0
        right_gain = max(get_max_gain(node.right), 0)
        best[0] = max(best[0], node.val + left_gain + right_gain)
        return node.val + max(left_gain, right_gain)

    get_max_gain(root)
    return best[0]


def build_tree(values):
  if not values or values[0]==None:
    return None
  root=TreeNode((values[0])) #5
  queue=deque([root]) #[5]
  i=1
  while queue and i<len(values):
    node=queue.popleft()#5
    if values[i]is not None:
      node.left=TreeNode((values[i]))
      queue.append(node.left)
    i+=1
    if i<len(values) and values[i]is not None:
      node.right=TreeNode((values[i]))
      queue.append(node.right)
    i+=1
  return root

print(max_path_sum(build_tree([2,-1])))

````

## Hard 5 — Match Parentheses with Wildcards

Given a string with `(`, `)`, and `*` (wildcard: can be `(`, `)`, or empty), return `True` if it can be a valid parenthesis string.

**Example:**
```
Input:  s = "(*)"
Output: True

Input:  s = "(*))"
Output: True

Input:  s = "(((*"
Output: False
```

````python
def check_valid_string(s: str) -> bool:
    # Track range of possible open-paren counts
    lo = hi = 0
    for ch in s:
        if ch == '(':
            lo += 1; hi += 1
        elif ch == ')':
            lo -= 1; hi -= 1
        else:  # '*'
            lo -= 1; hi += 1
        if hi < 0:
            return False
        lo = max(lo, 0)
    return lo == 0

````
