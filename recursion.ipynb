## 1. Factorial Calculation (Basic Linear Recursion)
**Problem Statement:** Write a recursive function that accepts a non-negative integer `n` and returns its mathematical factorial ($n!$).
* **Input:** `5`
* **Output:** `120` ($5 \times 4 \times 3 \times 2 \times 1$)

````python
def calculate_factorial(n: int) -> int:
    # Edge Case Guard: Intercept invalid negative inputs
    if n < 0:
        return -1

    # Base Case: The factorial of 0 is explicitly defined as 1
    if n == 0:
        return 1

    # Recursive Step: Multiply current number by the factorial of (n - 1)
    # This steadily shrinks the input parameter toward the base case anchor
    return n * calculate_factorial(n - 1)

# Test Cases
print("Factorial of 5:", calculate_factorial(5))    # Output: 120
print("Factorial of 0:", calculate_factorial(0))    # Output: 1
````

### Complexity Analysis
* **Time Complexity:** $O(N)$ — The function executes exactly $N$ sequential frames before terminating at the base case.
* **Space Complexity:** $O(N)$ — The system call stack must concurrently store $N$ frames before it can begin unwinding and calculating the multiplication chain backward.



2. The Call Stack
Explanation: The Call Stack is a hidden memory structure managed behind the scenes by the computer operating system. Because your system's processor can only handle one operational step at a time, it requires a strict organizer to remember exactly where it was when a parent function pauses to call a child subroutine.

The Call Stack operates strictly on a LIFO (Last-In, First-Out) principle. Each time a function calls itself, its local variables and current line of code are wrapped into a memory block called a Stack Frame and pushed onto the top of the stack. Execution halts on that line until the top frame encounters the base case and pops off, sending its result down to unleash the waiting frames beneath.

3. Multiple Recursion
Explanation: Multiple recursion occurs when a singular execution frame fires two or more completely separate, independent calls to itself. Instead of moving down in a clean, linear chain, multiple recursion branches outwards aggressively, creating a massive execution tree structure in the system memory.

The Golden Execution Rule: The computer never evaluates these multiple branches at the exact same moment. When the program hits the first recursive call on the line, the entire parent frame freezes. The processor dives entirely down that left-side branch until it runs completely dry, hits its base cases, and unwinds back to the parent. Only then does the processor shift horizontally to activate the right-side recursive call branch.

## 2. $n$-th Fibonacci Number (Naive Multiple Recursion)
**Problem Statement:** Implement a recursive function to find the $n$-th number in the Fibonacci sequence, where $F(0) = 0$ and $F(1) = 1$.
* **Input:** `6`
* **Output:** `8` (Sequence: 0, 1, 1, 2, 3, 5, 8...)

````python
def calculate_fibonacci(n: int) -> int:
    # Edge Case Guard: Intercept invalid negative indices
    if n < 0:
        return -1

    # Combined Base Cases: F(0) = 0, F(1) = 1
    if n <= 1:
        return n

    # Multiple Recursion Step: Spawns two distinct tracking branches
    # The processor executes the left branch fully before touching the right branch
    left_branch = calculate_fibonacci(n - 1)
    right_branch = calculate_fibonacci(n - 2)

    return left_branch + right_branch

# Test Cases
print("6th Fibonacci Number:", calculate_fibonacci(6))  # Output: 8
print("0th Fibonacci Number:", calculate_fibonacci(0))  # Output: 0
````

## Question 2: Flatten a Nested Dictionary (Call Stack Navigation)
**Problem Statement:** Write a recursive function that takes a deeply nested dictionary and flattens it into a single-level dictionary. Separate nested keys with a dot (`.`).
* **Input:** `{"a": 1, "b": {"c": 2, "d": {"e": 3}}}`
* **Output:** `{"a": 1, "b.c": 2, "b.d.e": 3}`

````python
def flatten_dict(nested_dict: dict, parent_key: str = '') -> dict:
    flat_dict = {}

    for key, value in nested_dict.items():
        # Formulate the new dot-separated key name
        new_key = f"{parent_key}.{key}" if parent_key else key

        if isinstance(value, dict):
            # Recursive Step: If the value is another dictionary, dive deeper
            # The Call Stack keeps track of our current flat_dict while we process sub-dicts
            flat_dict.update(flatten_dict(value, new_key))
        else:
            # Base Case: If it's a primitive value, assign it directly
            flat_dict[new_key] = value

    return flat_dict

# Test Case
nested = {"a": 1, "b": {"c": 2, "d": {"e": 3}}}
print("Flattened Dictionary:", flatten_dict(nested))
# Output: {'a': 1, 'b.c': 2, 'b.d.e': 3}
````

* **Time Complexity:** $O(N)$ where $N$ is the total number of key-value pairs across all levels of the dictionary.
* **Space Complexity:** $O(D)$ where $D$ is the maximum depth of nesting, dictating the maximum call stack height.


## Question 3: Binary Search Implementation (Basic Single-Branch Recursion)
**Problem Statement:** Implement a recursive binary search on a strictly sorted array to find the target's index. If the target does not exist, return `-1`.
* **Input:** `arr = [1, 3, 5, 7, 9, 11]`, `target = 9`
* **Output:** `4`

````python
def binary_search_recursive(arr: list, target: int, left: int, right: int) -> int:
    # Base Case 1: Elements exhausted, target not found
    if left > right:
        return -1

    mid = left + (right - left) // 2

    # Base Case 2: Target found at middle index
    if arr[mid] == target:
        return mid

    # Recursive Step (Single-branch choice)
    if arr[mid] > target:
        return binary_search_recursive(arr, target, left, mid - 1)
    else:
        return binary_search_recursive(arr, target, mid + 1, right)

# Test Case
nums = [1, 3, 5, 7, 9, 11]
print("Binary Search Target Index:", binary_search_recursive(nums, 9, 0, len(nums) - 1))  # Output: 4
````

* **Time Complexity:** $O(\log N)$ because the search space is divided in half with each individual call frame.
* **Space Complexity:** $O(\log N)$ storage required on the call stack due to single-line branching execution.


## Question 4: Count Steps (Multiple Recursion)
**Problem Statement:** You are climbing a staircase that takes `n` total steps to reach the top. Each time you can either climb `1` or `2` steps. Write a function to calculate how many distinct ways you can climb to the top.
* **Input:** `3`
* **Output:** `3` (Ways: $1+1+1$, $1+2$, $2+1$)

````python
def climb_stairs(n: int) -> int:
    # Base Case 1: 0 steps means you are already at the top (1 valid way found)
    if n == 0:
        return 1
    # Base Case 2: Negative steps mean you overshot the goal (0 valid ways)
    if n < 0:
        return 0

    # Multiple Recursion Step: Sum the path branching from a 1-step move
    # with the path branching from a 2-step move.
    return climb_stairs(n - 1) + climb_stairs(n - 2)

# Test Case
print("Distinct ways to climb 3 steps:", climb_stairs(3))  # Output: 3
print("Distinct ways to climb 4 steps:", climb_stairs(4))  # Output: 5
````

* **Time Complexity:** $O(2^N)$ representing exponential tree generation. Every level duplicates calls redundantly.
* **Space Complexity:** $O(N)$ depth constraint. The call stack will never exceed the height of the deepest individual linear branch ($n$).

## Question 5: Generate Balanced Parentheses (Multiple Branch Tracking)
**Problem Statement:** Given `n` pairs of parentheses, write a recursive function to generate all combinations of well-formed, balanced parentheses.
* **Input:** `2`
* **Output:** `["(())", "()()"]`

````python
def generate_parentheses(n: int) -> list:
    result = []

    def backtrack(current_str, open_count, close_count):
        # Base Case: When the string reaches its full valid length
        if len(current_str) == 2 * n:
            result.append(current_str)
            return

        # Branch 1: If we still have open parentheses available, add one
        if open_count < n:
            backtrack(current_str + "(", open_count + 1, close_count)

        # Branch 2: If there are more open than close parentheses, add a closing one
        if close_count < open_count:
            backtrack(current_str + ")", open_count, close_count + 1)

    # Start the execution tree with an empty string and 0 tracking counters
    backtrack("", 0, 0)
    return result

# Test Case
print("Balanced parentheses variants (n=2):", generate_parentheses(2))  # Output: ['(())', '()()']
````

* **Time Complexity:** $O(\frac{4^n}{\sqrt{n}})$ bounded by the Catalan numbers sequence, which defines valid bracket patterns.
* **Space Complexity:** $O(n)$ maximum call stack space to track a single valid path execution frame down to its leaf node.

## Question 6: Reverse a Stack Using Recursion (Call Stack Manipulation)
**Problem Statement:** Given a standard list representing a stack, reverse its elements entirely in-place using **only recursion**. You are not allowed to use loops or any auxiliary data structures.
* **Input:** `[1, 2, 3, 4]` (where 4 is the top of the stack)
* **Output:** `[4, 3, 2, 1]`

````python
def insert_at_bottom(stack: list, item) -> None:
    # Base Case: If stack is empty, place the item down
    if not stack:
        stack.append(item)
        return

    # Hold element to clear path to the bottom frame
    top_element = stack.pop()
    insert_at_bottom(stack, item)

    # Push back the held element on the way up
    stack.append(top_element)

def reverse_stack(stack: list) -> None:
    # Base Case: An empty stack is already reversed
    if not stack:
        return

    # Step 1: Pop the top element and hold it in the call stack frame
    top_element = stack.pop()

    # Step 2: Recursively empty the rest of the stack
    reverse_stack(stack)

    # Step 3: Insert the held element at the absolute bottom of the reversed stack
    insert_at_bottom(stack, top_element)

# Test Case
my_stack = [1, 2, 3, 4]
reverse_stack(my_stack)
print("Reversed Stack:", my_stack)  # Output: [4, 3, 2, 1]
````

* **Time Complexity:** $O(N^2)$ because `reverse_stack` calls itself $N$ times, and for each call, it invokes `insert_at_bottom` which can take up to $O(N)$ operations.
* **Space Complexity:** $O(N)$ system memory space required for the concurrent call stack frames.

## Question 7: Subsets / Power Set Generation (Multiple Recursion)
**Problem Statement:** Given an array of unique integers, return all possible subsets (the power set). The solution set must not contain duplicate subsets.
* **Input:** `[1, 2]`
* **Output:** `[[], [1], [2], [1, 2]]`

````python
def generate_subsets(nums: list) -> list:
    result = []

    def backtrack(index, current_subset):
        # Base Case: When we have made a decision for every element
        if index == len(nums):
            result.append(list(current_subset))
            return

        # Branch 1: Include the current element in the subset
        current_subset.append(nums[index])
        backtrack(index + 1, current_subset)

        # Branch 2: Exclude the current element (Backtrack)
        current_subset.pop()
        backtrack(index + 1, current_subset)

    backtrack(0, [])
    return result

# Test Case
print("All subsets for [1, 2]:", generate_subsets([1, 2]))  # Output: [[1, 2], [1], [2], []]
````

* **Time Complexity:** $O(N \cdot 2^N)$ where $2^N$ is the total number of subsets generated, and copying each subset into the result array takes $O(N)$ time.
* **Space Complexity:** $O(N)$ representing the maximum depth of the recursive tree execution frame stack.


## Question 8: Flood Fill Algorithm (Multi-Branch Matrix Traversal)
**Problem Statement:** Imagine an image represented by a 2D grid of integers. Given a starting coordinate `(sr, sc)`, a `new_color`, and a target color (the color at the start coordinate), change the color of the starting pixel, plus any pixels connected 4-directionally to it that share the same initial color.
* **Input:** `image = [[1,1,1],[1,1,0],[1,0,1]]`, `sr = 1`, `sc = 1`, `new_color = 2`
* **Output:** `[[2,2,2],[2,2,0],[2,0,1]]`

````python
def flood_fill(image: list[list[int]], sr: int, sc: int, new_color: int) -> list[list[int]]:
    rows, cols = len(image), len(image[0])
    target_color = image[sr][sc]

    # Edge case: If the target pixel is already the new color, return immediately
    if target_color == new_color:
        return image

    def dfs(r, c):
        # Base Case: Out of grid bounds or pixel is not the color we want to change
        if r < 0 or r >= rows or c < 0 or c >= cols or image[r][c] != target_color:
            return

        # Update current pixel color
        image[r][c] = new_color

        # Multiple Recursion: Fire 4 independent calls (Up, Down, Left, Right)
        dfs(r + 1, c) # Down
        dfs(r - 1, c) # Up
        dfs(r, c + 1) # Right
        dfs(r, c - 1) # Left

    dfs(sr, sc)
    return image

# Test Case
grid = [[1, 1, 1], [1, 1, 0], [1, 0, 1]]
print("Image Grid After Flood Fill:", flood_fill(grid, 1, 1, 2))
# Output: [[2, 2, 2], [2, 2, 0], [2, 0, 1]]
````

* **Time Complexity:** $O(M \cdot N)$ where $M$ is the number of rows and $N$ is the number of columns, as we visit every pixel at most once.
* **Space Complexity:** $O(M \cdot N)$ for the call stack in the worst-case scenario where the entire image needs to be filled.


## Question 9: Merge Two Sorted Linked Lists (Basic Recursion)
**Problem Statement:** Given the heads of two sorted linked lists, merge them into one continuous sorted linked list recursively.
* **Input:** `List1: 1 -> 3 -> 5`, `List2: 2 -> 4 -> 6`
* **Output:** `1 -> 2 -> 3 -> 4 -> 5 -> 6`

````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def merge_two_lists(l1: ListNode, l2: ListNode) -> ListNode:
    # Base Cases: If either list runs dry, return the remaining structural chain
    if not l1:
        return l2
    if not l2:
        return l1

    # Recursive Step: Compare values and link the smaller one
    if l1.val <= l2.val:
        l1.next = merge_two_lists(l1.next, l2)
        return l1
    else:
        l2.next = merge_two_lists(l1, l2.next)
        return l2

# Helper function to convert list to array to print inside Colab
def linked_list_to_list(head):
    res = []
    while head:
        res.append(head.val)
        head = head.next
    return res

# Test Case Setup
list1 = ListNode(1, ListNode(3, ListNode(5)))
list2 = ListNode(2, ListNode(4, ListNode(6)))
merged_head = merge_two_lists(list1, list2)
print("Merged Sorted Linked List:", linked_list_to_list(merged_head))
# Output: [1, 2, 3, 4, 5, 6]
````

* **Time Complexity:** $O(M + N)$ where $M$ and $N$ represent the respective lengths of the two sorted linked lists.
* **Space Complexity:** $O(M + N)$ auxiliary stack frame spaces utilized to resolve the recursive chain allocations.

## Question 10: Tower of Hanoi (Multiple Recursion)
**Problem Statement:** Move all disks from the source rod (`A`) to the destination rod (`C`) using an auxiliary rod (`B`). Larger disks can never sit on top of smaller disks.
* **Input:** `n = 2`
* **Output:** Moves printed sequentially.

````python
def tower_of_hanoi(n: int, source: str, destination: str, auxiliary: str) -> None:
    # Base Case: If there is only 1 disk left, move it directly
    if n == 1:
        print(f"Move disk 1 from {source} to {destination}")
        return

    # Step 1: Move top n-1 disks from Source to Auxiliary using Destination as a temporary hold
    tower_of_hanoi(n - 1, source, auxiliary, destination)

    # Step 2: Move the remaining largest disk directly to the Destination
    print(f"Move disk {n} from {source} to {destination}")

    # Step 3: Move the n-1 disks from Auxiliary to Destination using Source as a temporary hold
    tower_of_hanoi(n - 1, auxiliary, destination, source)

# Test Case
print("Tower of Hanoi Moves for 2 disks:")
tower_of_hanoi(2, 'A', 'C', 'B')
````

* **Time Complexity:** $O(2^N)$ as the number of operations strictly doubles with each added disk.
* **Space Complexity:** $O(N)$ stack frame overhead layout tracking the height depth of the current move operations.

**Recursion Use-cases**

Recursion is a powerful technique in programming that involves a function calling itself. In JavaScript, recursion can be used in various scenarios to solve problems.
Here are some of the common use cases of recursion

**Traversing Tree Structures:** Recursion is often used to traverse tree structures like JavaScript's Document Object Model (DOM). This makes it easy to navigate through complex nested structures.

Explanation: A recursive function can be used to traverse a tree by calling itself on each child node of the current node. This process continues recursively until all nodes in the tree have been visited. By using recursion, the code can be written in a more concise and readable manner compared to traditional iterative methods. Additionally, recursion can handle arbitrary levels of nesting within the tree, which can be difficult to handle with iterative methods.

**Sorting Algorithms: **Recursion can be used to implement sorting algorithms like Quick sort and Merge sort. These algorithms involve dividing a list into smaller sub-lists and sorting them recursively.

Explanation: In quicksort, the algorithm chooses a "pivot" element and divides the list into two sub-lists, one containing elements smaller than the pivot and one containing elements greater than the pivot. This process recursively occurs on the two sub-lists until the list is sorted.
In Merge sort, the algorithm divides the list into two sub-lists of roughly equal size. These sub-lists are recursively sorted and merged to form the final list.

**Searching and Filtering Arrays:** Recursion can be used to search and filter arrays. This involves calling the function with a smaller sub-array until the desired element is found or the array is empty.

Explanation: A binary search algorithm can be implemented recursively by dividing the array in half and checking if the middle element is the target element. If the middle element is not the target, the algorithm can call itself with either the left or right half of the array, depending on whether the target is greater or less than the middle element. This process recursively until the target element is found or the array is empty.

Similarly, a filter function can be implemented recursively by checking if the first element of the array satisfies a given condition. If it does, the element is added to a new array. The function is then called recursively with the rest of the array until the entire array is processed. The final result is the new array containing all the elements that satisfy the given condition.

**Generating Permutations and Combinations:** Recursion can be used to generate permutations and combinations of a set of elements. This involves calling the function with smaller elements until all possible permutations or combinations are generated.

Explanation: For permutations, the recursive function can be called with a smaller set of elements by removing one element at a time and generating all possible permutations of the remaining elements. The base case occurs when the set of elements is empty. Once a permutation is generated, it can be added to a list of permutations.
For combinations, the recursive function can be called with a smaller set of elements by either including or excluding the first element of the set. If the first element is included, the function is called recursively with the remaining elements and a smaller number of elements to choose from. If the first element is excluded, the function is called recursively with the remaining elements and the same number of elements to choose from. The base case occurs when the number of elements to choose is zero. Once a combination is generated, it can be added to a list of combinations.

**Solving Maze Problems:** Recursion can be used to solve maze problems. This involves calling the function recursively to explore all possible paths until the end of the maze is reached.
A maze Problem? It is a problem where the objective is to find a path from a starting point to an endpoint in a maze. The maze is typically represented as a grid of cells, where some cells are blocked and cannot be traversed. The path must traverse only unblocked cells and may not cross over itself. Recursion can be used to explore all possible paths in the maze until the endpoint is reached.

Explanation: The function can be designed to explore all possible paths by checking each adjacent cell to see if it is valid. The function can be called recursively with the new cell if it is a valid path. This process continues recursively until the end of the maze is reached, or all paths have been explored. By using recursion, the code can be written in a more concise and readable manner compared to traditional iterative methods.
Additionally, recursion allows backtracking to explore other possible paths, which can be challenging to handle with iterative methods.

**Parsing and Evaluating Expressions**: Recursion can be used to parse and evaluate expressions. This involves breaking down the expression into smaller sub-expressions and evaluating them recursively.

Explanation: Simple arithmetic expressions like 3 + 4 * 5 can be evaluated recursively by breaking them down into smaller sub-expressions like 4 * 5 and 3 + 20.

The recursive function can then be called with each sub-expression until the entire expression is evaluated. The base case occurs when the expression is a single value or variable, at which point the function returns the value of the expression. Using recursion, the code for parsing and evaluating expressions can be written more concisely and readable than traditional iterative methods.
Additionally, recursion can handle arbitrarily complex expressions, which can be challenging with iterative methods.

**When not to use Recursion**

Recursion is a powerful programming technique, but there are better solutions to a problem.

Here are five situations where using recursion is not recommended:

**When the input size is large:** Recursion requires a lot of memory overhead due to the call stack. If the input size is large, the call stack could overflow and cause a stack overflow error. For example, calculating the factorial of a large number using recursion could cause a stack overflow error.

**When the recursion depth is too deep**: Recursion requires a lot of stack space. If the recursion depth is too deep, then it could cause a stack overflow error. For example, calculating the Fibonacci sequence using recursion could cause a stack overflow error if the recursion depth is too deep.

**When the problem can be solved iteratively:** Recursion is not always the most efficient solution to a problem. In some cases, an iterative solution may be more efficient. For example, searching for an element in a sorted array can be done more efficiently using binary search, which is an iterative algorithm.
