### Optimized Solution

```python
def solve(s):
    # Base case: if the string is empty or a single character, it's already reversed.
    if len(s) <= 1:
        return s

    # Recursive case:
    # Take the last character: s[-1]
    # Append the result of reversing the rest of the string: s[:-1]
    return s[-1] + solve(s[:-1])

```

### Time Complexity

O(N) due to the maximum depth of the recursion stack, where N is the length of the string.

### Why This Solution Works
The function reduces the problem size by 1 at each step, moving closer to the base case. It builds the reversed string as the recursion unwinds.

### Optimized Solution

```python
def solve(nums):
    n = len(nums)
    # Initialize result array of the same size
    result = [0] * n

    # Two pointers at the ends of the input array
    left = 0
    right = n - 1

    # Pointer to fill the result array from the end (largest values first)
    index = n - 1

    while left <= right:
        # Compare absolute values. The larger absolute value will yield the larger square.
        if abs(nums[left]) > abs(nums[right]):
            result[index] = nums[left] **2
            left += 1
        else:
            result[index] = nums[right] **2
            right -= 1
        index -= 1

    return result

# Example
# Input: [-4,-1,0,3,10]
# Output: [0, 1, 9, 16, 100]
print(solve([-4,-1,0,3,10]))
```

### Time Complexity
O(N) where N is the length of `nums`. We process each element exactly once.

### Space Complexity
O(N) to store the `result` array.

### Why This Solution Works
Since the array is sorted, the numbers with the largest magnitude are always at the edges (either large negative numbers at the start or large positive numbers at the end). By comparing the elements at the two pointers, we can construct the result array starting from the largest square.

## Find All Anagrams in a String

### Difficulty
Easy

### Problem Statement
Given two strings `s` and `p`, return an array of all the start indices of `p`'s anagrams in `s`. You may return the answer in any order.

### Input Constraints
- `1 <= s.length, p.length <= 3 * 10^4`
- `s` and `p` consist of lowercase English letters.

### Example Input
s = \"cbaebabacd\", p = \"abc\"\n
### Example Output
[0, 6]

### Notes
Use a sliding window of size `len(p)` and frequency counters to compare the current window against `p`.

```python
# Write your solution here
def solve(s, p):
    pass

# Test your code below
```

### Optimized Solution

```python
def solve(s, p):
    # If p is longer than s, no anagrams can exist
    if len(p) > len(s):
        return []

    # Use frequency arrays since we only have lowercase English letters
    p_count = [0] * 26
    s_count = [0] * 26
    result = []

    # Initialize counts for the first window
    for i in range(len(p)):
        p_count[ord(p[i]) - ord('a')] += 1
        s_count[ord(s[i]) - ord('a')] += 1

    # Check the first window
    if p_count == s_count:
        result.append(0)

    # Slide the window across the rest of string s
    for i in range(len(p), len(s)):
        # Add the new character entering the window
        s_count[ord(s[i]) - ord('a')] += 1
        # Remove the character leaving the window
        s_count[ord(s[i - len(p)]) - ord('a')] -= 1

        # Check if the current window is an anagram
        if p_count == s_count:
            # The start index is current index i minus window size plus 1
            result.append(i - len(p) + 1)

    return result

# Example
# Input: s = \"abab\", p = \"ab\"\n# Output: [0, 1, 2]
```

### Time Complexity
O(S + P) where S and P are lengths of strings `s` and `p`. Array comparisons take O(26) which simplifies to O(1).

### Space Complexity
O(1) because the frequency arrays are always size 26, independent of input size.

### Why This Solution Works
Comparing substrings directly or sorting every window would be too slow. By maintaining a running frequency count (sliding window), we can update the state in O(1) time and perform an O(1) comparison against the target frequency.

### Optimized Solution

```python
def solve(x, n):
    # Base case: anything to the power of 0 is 1
    if n == 0:
        return 1.0
    # Handle negative powers
    if n < 0:
        return 1.0 / solve(x, -n)
    # Recursive case: divide and conquer
    half = solve(x, n // 2)
    # If n is even, (x^(n/2))^2
    if n % 2 == 0:
        return half * half
    # If n is odd, x * (x^(n//2))^2
    else:
        return half * half * x

# Example
# Input: x = 2.0, n = 3
# Output: 8.0
```

### Time Complexity
O(log n). The exponent is halved at each recursive step, logarithmically reducing the problem size.

### Space Complexity
O(log n) for the recursion stack.

### Why This Solution Works
Instead of multiplying `x` by itself `n` times, we calculate `x^(n/2)` once and square it. This cuts the number of operations drastically, preventing recursion depth limits.

### Optimized Solution

```python
def solve(nums, target):
    def find_bound(is_first):
        left, right = 0, len(nums) - 1
        ans = -1
        while left <= right:
            mid = left + (right - left) // 2
            if nums[mid] > target:
                right = mid - 1
            elif nums[mid] < target:
                left = mid + 1
            else:
                # Target found. Record the position.
                ans = mid
                if is_first:
                    # If searching for the first occurrence, keep searching left
                    right = mid - 1
                else:
                    # If searching for the last occurrence, keep searching right
                    left = mid + 1
        return ans

    first_pos = find_bound(True)
    # Optimization: If the first position isn't found, the element isn't in the array.
    if first_pos == -1:
        return [-1, -1]

    last_pos = find_bound(False)
    return [first_pos, last_pos]

# Example
# Input: nums = [5,7,7,8,8,10], target = 6
# Output: [-1, -1]
print(solve([1,2,2,3], 2))
```

### Time Complexity
O(log N). We run binary search at most twice, each taking O(log N) time.

### Space Complexity
O(1) as we only use a few integer variables.

### Why This Solution Works
Standard binary search stops when it finds the target. By modifying the condition when `nums[mid] == target`, we force the search to continue either entirely in the left half (to find the start) or entirely in the right half (to find the end).

This requires backtracking. Maintain a `path` of current elements and an array of `visited` elements, or swap elements in place.

```python
# Write your solution here
def solve(nums):
    pass

# Test your code below
```

### Optimized Solution

```python
def solve(nums):
    res = []

    def backtrack(path, used):
        # Base case: if the current permutation has the same length as nums, it is complete.
        if len(path) == len(nums):
            # Create a copy of the path and append to results
            res.append(path[:])
            return

        # Iterate through all available choices
        for i in range(len(nums)):
            # Pruning condition: skip elements already used in the current path
            if used[i]:
                continue

            # Make a choice
            path.append(nums[i])
            used[i] = True

            # Explore the branches following this choice
            backtrack(path, used)

            # Backtrack: undo the choice to explore other branches
            path.pop()
            used[i] = False

    # Start backtracking with an empty path and a boolean array tracking usage
    backtrack([], [False] * len(nums))
    return res

# Example
# Input: [0, 1]
# Output: [[0, 1], [1, 0]]
```

### Time Complexity
O(N * N!) where N is the length of `nums`. There are N! permutations, and for each, we copy the path which takes O(N) time.

### Space Complexity
O(N) for the recursion stack and the `path`/`used` arrays. We do not count the output array `res` in auxiliary space complexity.

### Why This Solution Works
Backtracking explores all possible choices. By using a boolean array to track which elements are currently in the `path`, we ensure elements aren't reused within the same permutation. The `pop()` and `False` reassignment perfectly restores the state for the next sibling branch.

For every element, you have two choices: include it in the current subset, or exclude it.

```python
# Write your solution here
def solve(nums):
    pass

# Test your code below
```

### Optimized Solution

```python
def solve(nums):
    res = []

    def backtrack(index, current_subset):
        # Base case: we have considered all elements
        if index == len(nums):
            # Add a copy of the current subset to results
            res.append(current_subset[:])
            return

        # Choice 1: Include the current element
        current_subset.append(nums[index])
        # Recurse to the next index
        backtrack(index + 1, current_subset)

        # Choice 2: Exclude the current element (Backtrack)
        current_subset.pop()
        # Recurse to the next index without the element
        backtrack(index + 1, current_subset)

    backtrack(0, [])
    return res

# Example
# Input: [0]
# Output: [[], [0]]
```

### Time Complexity
O(N * 2^N). There are 2^N possible subsets. At each base case, creating a copy of `current_subset` takes up to O(N) time.

### Space Complexity
O(N) for the maximum depth of the recursion stack and the space used by `current_subset`.

### Why This Solution Works
This uses standard backtracking to build a decision tree. Each node in the tree represents an index in `nums`, branching left if we include `nums[index]` and right if we exclude it. The leaves represent all possible subsets.

## Minimum Size Subarray Sum

### Problem Statement

### Input Constraints

- `1 <= nums.length <= 10^5`
- `1 <= nums[i] <= 10^4`

### Example Input
target = 7, nums = [2,3,1,2,4,3]

### Example Output
2

### Notes
Brute force checking all subarrays is O(N^2) and will TLE. Because all elements are positive, the window sum grows monotonically as it expands. Use a Sliding Window.

```python
# Write your solution here
def solve(target, nums):
    pass

# Test your code below
```

### Optimized Solution

```python
def solve(target, nums):
    # Initialize pointers and running sum
    left = 0
    current_sum = 0
    # Initialize min_length to infinity so any valid length will overwrite it
    min_length = float('inf')

    # Expand window by moving the right pointer
    for right in range(len(nums)):
        current_sum += nums[right]

        # While the window meets the condition, try to shrink it to find a smaller length
        while current_sum >= target:
            # Update minimum length found so far
            min_length = min(min_length, right - left + 1)
            # Subtract the element exiting the window from the left
            current_sum -= nums[left]
            # Shrink the window
            left += 1

    # If min_length wasn't updated, no valid subarray exists
    return min_length if min_length != float('inf') else 0

# Example
# Input: target = 4, nums = [1,4,4]
# Output: 1
```

### Time Complexity
O(N). Both the `left` and `right` pointers traverse the array at most once. The inner `while` loop advances the `left` pointer, meaning the total number of operations inside the loops is bounded by 2N.

### Space Complexity
O(1). We use a few primitive variables regardless of the input array size.

### Why This Solution Works
Since elements are strictly positive, adding an element increases the sum, and removing one decreases it. A sliding window leverages this. Once a valid window is found, we shrink it from the left until it becomes invalid, logging the minimal valid sizes.

## Combination Sum

### Difficulty

### Problem Statement
Given an array of distinct integers `candidates` and a target integer `target`, return a list of all unique combinations of `candidates` where the chosen numbers sum to `target`. You may return the combinations in any order.
The same number may be chosen from `candidates` an unlimited number of times. Two combinations are unique if the frequency of at least one of the chosen numbers is different.

### Example Input

### Example Output
[[2,2,3],[7]]

### Notes
This combines Backtracking with Pruning logic. Because you can reuse numbers, your recursive call shouldn't automatically advance the index. Prune branches where the sum exceeds the target.

```python
# Write your solution here
def solve(candidates, target):
    pass

# Test your code below
```

### Optimized Solution

```python
def solve(candidates, target):
    res = []
    # Optional: sorting helps prune earlier if an element exceeds target
    candidates.sort()

    def backtrack(start_index, current_sum, current_path):
        # Base case 1: Found a valid combination
        if current_sum == target:
            res.append(current_path[:])
            return

        # Iterate through candidates starting from 'start_index' to avoid permutations
        for i in range(start_index, len(candidates)):
            candidate = candidates[i]

            # Pruning condition: Since candidates are sorted, if adding the current
            # candidate exceeds target, subsequent ones will also exceed it.
            if current_sum + candidate > target:
                break

            # Choose the element
            current_path.append(candidate)

            # Recurse. Note we pass 'i' as start_index, NOT 'i + 1',
            # allowing the same element to be reused indefinitely.
            backtrack(i, current_sum + candidate, current_path)

            # Backtrack to explore other choices
            current_path.pop()

    backtrack(0, 0, [])
    return res

# Example
# Input: candidates = [2], target = 1
# Output: []
```

### Time Complexity
O(N^(T/M)) in the worst case, where N is the number of candidates, T is the target, and M is the minimum value among candidates. The depth of the recursion tree can go up to T/M. Sorting takes O(N log N).

### Space Complexity
O(T/M) for the recursion stack depth. The `current_path` list also takes up to O(T/M) space.

### Why This Solution Works
Backtracking explores all potential combinations. By passing `start_index` forward, we ensure we generate combinations `[2, 3]` without generating redundant permutations like `[3, 2]`. Sorting allows us to prune the search space aggressively: as soon as `current_sum + candidate > target`, we break the loop entirely.
