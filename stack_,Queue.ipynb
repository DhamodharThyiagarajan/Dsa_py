```python
class Stackimpl:
  def __init__(self):
    self.stack=[]

  def push(self,item):
    self.stack.append(item)
    print(f'item pushed {item}')

  def pop(self):
    if len(self.stack)==0:
      print('stack is empty')
      return
    self.stack.pop()
  def popelement(self,item):
    if len(self.stack)==0:
      print('stack is empty')
      return
    b=[]
    while len(self.stack)>=1:
      t=self.stack.pop()
      if t==item:
        while len(b)>=1:
          self.stack.append(b.pop())
        break
      b.append(t)

  def show(self):
    print(self.stack)
s=Stackimpl()
s.push(10)
s.push(20)
s.push(30)
s.push(40)
s.push(50)
s.popelement(30)
s.show()
```

```python
from collections import deque
class DequeueImpl:
  def __init__(self):
    self.queue=deque()

  def enqueue(self,item):
    self.queue.append(item)
    print(f'item enqueued {item}')

  def dequeue(self):
    if len(self.queue)==0:
      print('queue is empty')
      return
    self.queue.popleft()

  def show(self):
    print(self.queue)
d=DequeueImpl()
d.enqueue(10)
d.enqueue(20)
d.enqueue(30)
d.dequeue()
d.dequeue()
d.dequeue()
d.dequeue()
d.show()
```

```python
from collections import deque
def time_to_required_buy_tickets(tickets,k):
  queue=deque([i for i in range(len(tickets))])
  time=0
  while queue:
    current_person=queue.popleft()
    tickets[current_person]-=1
    time+=1
    if tickets[current_person]==0 and current_person==k:
      return time
    if tickets[current_person]>0:
      queue.append(current_person)
print(time_to_required_buy_tickets([2,3,2],2))
```
